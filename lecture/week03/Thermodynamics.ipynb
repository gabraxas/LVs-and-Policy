## 배기속도 — 어떻게 배기속도 Ve를 최대화할 것인가

- 배기속도를 어떻게 최대화할 것인가
- 이론적 최대 배기속도
  - 정압(c.v) 조건에서의 열역학 제1법칙
  - ${h}_{t2}={h}_{t1}+ q ={h}_{ e }+\frac{1}{2}{ U }_{ e }^{2}$
  - 최대 배기속도는 무한팽창 시 얻어짐(${h}_{e}=0$)
  - ${ U }_{ e , max } \sim \sqrt{2{h}_{t2}}=\sqrt{2{ c }_{ p }{ T }_{t2}}=\sqrt{\frac{2 \gamma }{ \gamma - 1} R { T }_{t2}}=\sqrt{\frac{2 \gamma }{ \gamma - 1}\left( \frac{{ R }_{ u }}{{ M }_{ w }} \right){ T }_{t2}}$
  
- 이론적 최대 배기속도는 연소실 온도와 배기가스 분자량에 의해 결정됨

- 등엔트로피 과정 복습
  - 열역학 법칙 정리
    - 열역학 제0법칙
      - A가 C와 열평형, B도 C와 열평형이면 A와 B도 열평형 상태여야 함
    - 열역학 제1법칙
      - 에너지 보존
      - 계는 내부에너지·열·일 세 종류의 에너지만을 가짐
      - $d e = \delta q - \delta w$
      - 단순 물질의 경우 일은 압력일(pressure work)뿐임
      - $\delta w = Pdv$
      - 엔탈피로 표현하면
      - $d h= \delta q + vdP$
        - $e$ : 내부에너지
        - $h$ : 엔탈피
        - $q$ : 열전달량
        - $w$ : 일량
        - $P$ : 압력
        - $v$ : 비체적
- 열역학 복습
    - 열역학 제2법칙
      - 외부 에너지원이 없다면 열은 항상 고온에서 저온으로 흐름
      - 단일성분계에서 다음과 같이 정의되는 상태변수 '엔트로피'가 존재함
      - $d s =\frac{ \delta q }{ T }$ 	$d S =\frac{ \delta Q }{ T }$
      - 열역학 제2법칙
      - $d { S }_{universe} \geq 0$
      - $d { S }_{system}+ d { S }_{surroundings} \geq 0$
      - 고립계의 경우
      - $d { S }_{system} \geq 0$
      - 엔트로피는 가능한 미시상태에 대한 '무질서도' 또는 '정보 부족'의 척도
      - $S = k \ln {\left( { N }_{micro} \right)}$
- $S$ : 엔트로피
- $s$ : 비엔트로피
- $T$ : 온도
  - 이상기체($P v = R T$)의 가역·단열 과정($d s =0$) 가정 시:
  - $\delta Q = d e + Pdv$
  - $Tds = d \left( h - P v \right)+ Pdv$
  - $Tds = d h - Pdv - vdP + Pdv$
  - $Tds = d h - vdP$
  - $d s =\frac{{ c }_{ p }}{ T } d T - \frac{ R }{ P } d P$
  - $0=\frac{{ c }_{ p }}{ T } d T - \frac{ R }{ P } d P$
  - $\int_{1}^{2} {\frac{{ c }_{ p }}{ T } d T } - \int_{1}^{2} {\frac{ R }{ P } d P }= const .$
  - $\ln {\left( \frac{{ T }_{2}}{{ T }_{1}} \right)} - \frac{ R }{{ c }_{ p }}\ln {\left( \frac{{ P }_{2}}{{ P }_{1}} \right)}= const .$
- 관을 통과하는 유동
  - 기체가 관을 통과할 때 기체 분자는 관 벽에 의해 편향됨
    - 유속이 음속보다 훨씬 느리면 밀도는 일정하게 유지되고 유속만 증가함
    - 유속이 음속에 근접하면 압축성 효과를 고려해야 함
    - [Ceteris Paribus] 유동이 매우 서서히 압축(단면적 감소)된 후 서서히 팽창(단면적 증가)하면, 유동 조건은 원래 상태로 되돌아옴
    - 이러한 과정을 가역과정이라 함(열역학 제2법칙)

- 먼저, 음속이란 무엇인가?
  - '음속'이란 매질을 통한 작은 교란의 전파 속도임
    - 교란은 기체 내 분자들의 무작위 운동에 의한 충돌을 통해 전파됨
    - 작은 교란의 전파는 등엔트로피 과정임
    - 전파 속도는 분자 충돌에 의존하므로, 음속은 기체의 상태에 따라 달라짐
    - [Ceteris Paribus] 비열이 일정한 기체(칼로리 완전기체)라고 가정하면
      - 질량보존과 운동량보존에서 출발하면:
      - $\dot{ m }= \rho u A$	$- \frac{ d P }{ d x }= \rho u\frac{ d u}{ d x }$
      - $\rho u = \rho a =\left( \rho + d \rho \right) \cdot ( a + d u )$
      - $\rho a = \rho a + \rho d u + a d \rho + d \rho d u$
      - $\rho d u = - a d \rho$
      - $d P ={ a }^{2} d \rho$
      - $a =\sqrt{\frac{ d P }{ d \rho }}$
        - $P$
        - $\rho$
      - $d P / d \rho ={ a }^{2}$

    - 기체의 엔트로피는:

    - $\frac{{ c }_{ p }}{ R } d \left( \frac{ P }{ \rho } \right)=\frac{ d P }{ \rho }$

    - $\left( \frac{1}{ \gamma - 1} - 1 \right)\frac{ d P }{ P }=\left( \frac{ \gamma }{ \gamma - 1} \right)\frac{ d \rho }{ \rho }$

      - $\gamma P \frac{ d \rho }{ \rho }={ a }^{2} d \rho$
      - $\frac{ \gamma P }{ \rho }={ a }^{2}$
    - 온도가 더 직관적인 상태변수이므로 이상기체 상태방정식($P = \rho R T$)을 이용해 $\rho$를 $T$로 치환하면
    - ${ a }^{2}= \gamma R T$

## 비열비란 무엇인가?


- 비열비
  - $\gamma$는 ${ c }^{ \ast }$와 ${ C }_{ F }$ 모두에 등장하며, 비열비의 정의는
  - $\gamma ={ c }_{ p }/{ c }_{ v }$
  - 왜 세상은 $\gamma$에 의해 지배되는가?
  - 이유 1) 등엔트로피 과정·이상기체 상태방정식에서, 정보는 음속 $a =\sqrt{ \gamma R T }$을 통해 전달됨
  - 이유 2) '온도'와 '기체의 에너지'의 정의로부터
    - 첫 번째 질문) 물질의 온도란 무엇인가?
    - 두 번째 질문) 내부에너지와 엔탈피란 무엇인가?
  - [Ceteris Paribus] 이상기체를 가정
    - 기체 분자의 부피를 무시
    - 완전탄성충돌(운동량 보존)
    - 분자 간 힘의 상호작용 없음
    - 분자의 운동은 뉴턴 법칙을 따르며 무작위적·비상관적임
  - 압력을 다음과 같은 거시적 물성으로 정의함
    - 단위면적당 분자가 벽에 충돌하며 전달하는 운동량 변화

![](https://raw.githubusercontent.com/gabraxas/LVs-and-Policy/main/lecture/week02/assets/s045_img002.gif)

  - 한 변의 길이가 L인 정육면체 용기에 질량 m인 단원자분자 N개가 들어있다고 가정
  - 벽에 작용하는 i번째 기체분자의 힘 = x축 운동량 변화량:
  - $\Delta { p }_{xi}=\left( - m { v }_{xi} \right) - \left( m { v }_{xi} \right)= - 2 m { v }_{xi}$
  - ${ F }_{xi}=\frac{\left| \Delta { p }_{xi} \right|}{ \Delta t }=\frac{2 m { v }_{xi}}{\frac{2 L }{{ v }_{ x }}}=\frac{ m { v }_{xi}^{2}}{ L }$
  - 여기서 벽 충돌 시간간격은 $\Delta t =2 L /{ v }_{ x }$
  - 전체 N개 기체분자가 벽에 작용하는 힘 F는:
  - ${ F }_{ x }=\sum_{ i =1}^{ N } {{ F }_{xi}}=\sum_{ i =1}^{ N } {\frac{\left| \Delta { p }_{xi} \right|}{ \Delta t }}=\frac{ m }{ L }\sum_{ i =1}^{ N } {{ v }_{xi}^{2}}=\frac{ m }{ L } N {\left( { v }_{ x }^{2} \right)}_{avg}$
  - 벽 하나에 작용하는 압력은:
  - $P =\frac{{ F }_{ x }}{{ L }^{2}}=\frac{ m }{{ L }^{3}} N {\left( { v }_{ x }^{2} \right)}_{avg}=\frac{ m }{ V }\left( n { N }_{ A } \right){\left( { v }_{ x }^{2} \right)}_{avg}$
  - $=\frac{ n }{ V } M {\left( { v }_{ x }^{2} \right)}_{avg}$
  - $=\frac{ n M }{ V }{\left( \frac{1}{3}{ v }^{2} \right)}_{avg}$
  - $=\frac{ n M }{3 V }({ v }_{rms}^{2})$
  - $P V = n \left( \frac{ M { v }_{rms}^{2}}{3} \right)$

![](https://raw.githubusercontent.com/gabraxas/LVs-and-Policy/main/lecture/week02/assets/s046_img002.jpg)

- ${ N }_{ A }$ : 아보가드로 수, ~6.02×10²³
  - $P$, $T$, $V$의 이상기체 상태방정식 $P V = nRT$를 이용하면
  - $\frac{ M { v }_{rms}^{2}}{3}= R T$
  - ${ v }_{rms}=\sqrt{\frac{3 R T }{ M }}$
  - 단위 이상기체 분자의 운동에너지는 병진운동 항으로 정의됨
  - ${ K }_{avg}={\left( \frac{1}{2} m { v }^{2} \right)}_{avg}=\frac{1}{2} m {\left( { v }^{2} \right)}_{avg}=\frac{1}{2} m { v }_{rms}^{2}$
  - 온도로 나타내면,
  - ${ K }_{avg}=\frac{1}{2} m \left( \frac{3 R T }{ M } \right)=\frac{3}{2}\left( \frac{ m }{ M } \right) R T =\frac{3}{2}\left( \frac{1}{{ N }_{ A }} \right) R T =\frac{3}{2}\left( \frac{ R }{{ N }_{ A }} \right) T$
  - 볼츠만 상수 ${ k }_{ B }= R /{ N }_{ A }$를 도입하면
  - ${ K }_{avg}=\frac{3}{2}{ k }_{ B } T$
  - 이 식은 온도 T에서 모든 기체분자가 질량에 무관하게 동일한 평균 운동에너지를 가진다는 것을 의미함
  - 앞서 유도한 ${ v }_{ x }^{2}={ v }^{2}/3$ 과정에서, 움직일 수 있는 자유도 f마다 평균 운동에너지 ${ K }_{avg}=1/2{ k }_{ B } T$를 갖는다고 해석할 수 있음
  - 즉 분자운동의 자유도 f를 알면, 분자의 운동에너지(=내부에너지)를 온도의 함수로 나타낼 수 있으며, 그 비례상수는 볼츠만 상수 ${ k }_{ B }$의 절반임

  - $\gamma =\frac{{ c }_{ p }}{{ c }_{ v }}=\frac{ f +2}{ f }$

  - 비열비가 크다는 것의 의미:
    - 열용량 증가 — 비열비가 높을수록 정압 조건에서 정적 조건 대비 더 많은 열에너지를 흡수함. 즉 단위질량(또는 몰)당 열에너지 저장 능력이 더 큼
    - 팽창 거동 — 비열비가 클수록 정압 가열 시 기체가 더 잘 팽창하는 경향을 보임. 정압 조건에서 온도를 올리는 데 정적 조건보다 더 많은 에너지가 필요하기 때문임
    - 단열과정 — 단열과정에서는 비열비가 높을수록 동일한 압력 변화에 대해 온도 변화가 더 커짐. 단열과정에서 온도-압력 관계가 비열비에 좌우되기 때문임
    - 음속 — 비열비가 클수록 일반적으로 기체 내 음속이 높아짐
    - 열역학 사이클의 효율 — 오토(Otto) 사이클, 브레이턴(Brayton) 사이클, 랭킨(Rankine) 사이클 등에서 비열비는 효율에 영향을 미침. 특정 조건에서는 비열비가 높을수록 열효율이 높아질 수 있음

## 특성속도(Characteristic Velocity)

- 특성속도 ${ c }^{ \ast }$
  - 초크(choked) 유동 상관식을 다시 살펴보면:
  - $\dot{ m }=\frac{{ A }_{ \ast }{ P }_{ t }}{\sqrt{{ T }_{ t }}}\sqrt{\frac{ \gamma }{ R }}{\left( \frac{ \gamma +1}{2} \right)}^{ - \frac{ \gamma +1}{2\left( \gamma - 1 \right)}}$	(1)
  - 추력($F$)의 정의로부터
  - $F =\dot{ m }{ V }_{ e }$	(2)
  - 압력 ${ P }_{ t }$의 챔버를 가정하면, 목(throat) ${ A }_{ \ast }$에 작용하는 힘은:
  - $F ={ P }_{ t }{ A }_{ \ast }$	(3)
  - (2)와 (3)을 결합해 노즐 목의 특성속도($M =1$)를 구함 — 이를 특성속도(${ c }^{ \ast }$)라 부름
  - $\dot{ m }{ V }_{ e }^{ \ast }={ P }_{ t }{ A }_{ \ast }$
  - ${ c }^{ \ast } \equiv { V }_{ e }^{ \ast }=\frac{{ P }_{ t }{ A }_{ \ast }}{\dot{ m }}$	(4)
  - 노즐 목의 질량유량을 대입하면
  - ${ c }^{ \ast }=\frac{\sqrt{ \gamma }}{ \gamma \sqrt{{\left( \frac{2}{ \gamma +1} \right)}^{\frac{ \gamma +1}{ \gamma - 1}}}}\sqrt{\frac{{ R }_{ u }}{{ M }_{ w }}{ T }_{ t }}$
  - 특성속도는 초음속 노즐 부분을 제외한, 연소와 화학에너지 수준이 추력 형성에 기여하는 정도를 나타내는 척도임
  - c*가 클수록 유효 배기속도가 높고, 성능이 높음을 의미!
- ${ M }_{ w }$ : 분자량
- ${ R }_{ u }$ : 기체상수
  - ${ c }^{ \ast }$에 유의미한 영향을 주는 인자는?
    - $f \left( \gamma \right) \equiv \frac{\sqrt{ \gamma }}{ \gamma \sqrt{{\left( \frac{2}{ \gamma +1} \right)}^{\frac{ \gamma +1}{ \gamma - 1}}}}$

## 추력계수(Thrust Coefficient)

- 추력계수 ${ C }_{ F }$
  - 추력 방정식 복습:
  - $T =\dot{ m }{ u }_{ e }+\left( { P }_{ e } - { P }_{ a } \right){ A }_{ e }$
  - 추력계수(실제 추력 대 초크노즐 추력의 비):
  - ${ C }_{ F } \equiv \frac{ T }{{ A }_{ \ast }{ P }_{t2}}$
  - ${ C }_{ F }=\frac{\dot{ m }{ u }_{ e }}{{ A }_{ \ast }{ P }_{t2}}+\left( \frac{{ P }_{ e }}{{ P }_{t2}} - \frac{{ P }_{ a }}{{ P }_{t2}} \right)\frac{{ A }_{ e }}{{ A }_{ \ast }}$
  - 등엔트로피·칼로리 완전기체 노즐 유동 가정 시 추력계수는 다음과 같이 표현됨
  - ${ C }_{ F }={\left\{ \left( \frac{2{ \gamma }^{2}}{ \gamma - 1} \right){\left( \frac{2}{ \gamma +1} \right)}^{\frac{ \gamma +1}{ \gamma - 1}}\left[ 1 - {\left( \frac{{ P }_{ e }}{{ P }_{t2}} \right)}^{\frac{ \gamma - 1}{ \gamma }} \right] \right\}}^{1/2}+\left( \frac{{ P }_{ e }}{{ P }_{t2}} - \frac{{ P }_{ a }}{{ P }_{t2}} \right)\frac{{ A }_{ e }}{{ A }_{ t }}$
- 노즐 내 충격파 유발 유동박리
  - 서머필드(Summerfield) 기준:
    - 박리 위험이 발생하는 조건
    - $\frac{{ P }_{t2}}{{ P }_{ a }}>16$, 	${ \alpha }_{ e } \sim {15}^{ o }$
    - 등엔트로피 유동 방정식으로 ${ P }_{ e }$를 계산함
    - 박리 발생 가능 조건
    - $\frac{{ P }_{ e }}{{ P }_{ a }}<{\left( \frac{{ P }_{ e }}{{ P }_{ a }} \right)}_{cr} \sim 0.40$
    - ${ \alpha }_{ e }>{15}^{ o }$이면 ${\left( \frac{{ P }_{ e }}{{ P }_{ a }} \right)}_{cr}$는 더 낮아질 수 있음
    - 대형 노즐의 경우 최신 데이터에 따르면
    - $\frac{{ P }_{ e }}{{ P }_{ a }}<{\left( \frac{{ P }_{ e }}{{ P }_{ a }} \right)}_{cr} \sim 0.286$
    - 박리 유동은 경사충격파 시작점 이후의 확산부를 무시함
    - 이는 추력계수 저하폭을 제한함

## 비추력(Specific Impulse)


- 특성속도(${ c }^{ \ast }$)와 추력계수(${ C }_{ F }$)의 정의를 결합해 추력을 표현하면
  - ${ c }^{ \ast }=\frac{{ P }_{t2}{ A }_{ \ast }}{\dot{ m }}$,	${ C }_{ F }=\frac{ T }{{ A }_{ \ast }{ P }_{t2}}$
  - $T ={ C }_{ F }{( P }_{t2}{ A }_{ \ast })$
  - $T =\dot{ m }$ $\times \mathop{\overbrace{{ c }^{ \ast }}}\limits_{챔버 내 기체의 에너지} \times \mathop{\overbrace{{ C }_{ F }}}\limits_{노즐 내 유동}$
  - 비추력 : 분사된 단위질량당 추력
  - ${ I }_{sp} \equiv \frac{ T }{\dot{ m }{ g }_{0}}=\frac{{ c }^{ \ast }{ C }_{ F }}{{ g }_{0}}$
  - 임펄스밀도 : 분사된 단위체적당 추력
  - $\delta \equiv \frac{ T }{{\dot{ V }}_{ p }}=\frac{ T }{\dot{ m }{ g }_{0}}\frac{\dot{ m }{ g }_{0}}{{\dot{ V }}_{ p }}={ I }_{sp}{ \rho }_{ p }$
  - 총 임펄스
  - ${ I }_{tot}=\int_{0}^{{ t }_{ b }} { Tdt }=\int_{0}^{{ t }_{ b }} {\dot{ m }{ I }_{sp}{ g }_{0} d t }$
  - 평균 추력
  - $\overline{ T } \equiv \frac{1}{{ t }_{ b }}\int_{0}^{{ t }_{ b }} { Tdt =\frac{{ I }_{tot}}{{ t }_{ b }}}$
