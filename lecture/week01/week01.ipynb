### 1.2 16주 로드맵

$$
\text{기술 기초(2~5주)} \;\to\; \text{정책·역사(6~10주)} \;\to\; \text{비즈니스·시장(11~13주)} \;\to\; \text{한국·종합(14~16주)}
$$

| 구간 | 단계 | 내용 |
|---|---|---|
| 2~5주 | 기술 기초 | 로켓 이론 · 추진 · 시스템 설계 · 재사용 경제성 |
| 6~10주 | 정책·역사 | 우주법·발사허가·보험(7주) · 미국/각국 정책 — 8주 팀 확정 |
| 11~13주 | 비즈니스·시장 | BMC 사업전략(11) · TRANSCOST 시장경제학(12) · 스타십 시나리오(13) |
| 14~16주 | 한국·종합 | 한국 정책+중간발표(14) · 최종발표·협상 역할극(15) · 기말(16) |

1. 전반부: 기술의 정량 직관 확보 (가정 없는 사업계획 = 공상)
2. 중반부: 기술이 놓인 제도(허가·보험)와 정책(각국 육성 논리) 결합
3. 후반부: 비즈니스 도구(BMC·TRANSCOST·시나리오)로 종합 → 팀별 사업기획보고서 완성

## 2. 우주수송이란 무엇인가

### 2.1 정의와 가치사슬 속 위치

- 우주수송 = 탑재체를 지상에서 목표 궤도·천체까지 운반하는 서비스 + 그 수단(발사체)
- 우주경제 가치사슬에서 발사서비스 = **'업스트림의 관문'**

| 구간 | 내용 | 비고 |
|---|---|---|
| 업스트림 (제조) | 발사체·위성·부품 제조 | 발사서비스 시장 ~\$14\~18B |
| **미드스트림 (수송·운용)** | **발사서비스 · 지상국 · 관제** | **← 이 강좌의 대상** |
| 다운스트림 (서비스) | 통신·항법·관측 서비스와 응용 | 우주경제의 대부분 (예: Starlink 매출 ~\$11B/년) |

> **핵심 명제**: 2025년 세계 우주경제 \~\$626B 중 발사서비스는 \~\$14\~18B(2~3%). 그런데도 각국이 발사체에 국력을 거는 이유 — 나머지 97%가 전부 이 관문을 통과해야 하고, 관문의 가격·가용성·주권이 다운스트림 전체의 조건을 결정하기 때문.

### 2.2 기본 용어: 궤도 · 체급 · $/kg

**궤도**

| 궤도 | 고도·특징 | 주 수요 |
|---|---|---|
| LEO (저궤도) | ~200–2,000 km — 가장 낮은 에너지 | 군집통신(Starlink 등), 관측, ISS |
| SSO (태양동기) | ~500–800 km 극궤도 — 동일 지방시 관측 | 지구관측 위성의 표준 궤도 |
| GTO/GEO (정지) | ~35,786 km — 지구 자전 동기 | 대형 통신·방송, 기상 |

**체급 (LEO 기준)**

| 체급 | 탑재능력 | 예시 |
|---|---|---|
| 소형 (Small) | ~2 t 이하 | Electron, 한국 스타트업들 |
| 중형 (Medium) | 2~20 t | Falcon 9, 누리호(1.9t)·KSLV-III |
| 대형 (Heavy) | 20~50 t | Falcon Heavy, New Glenn |
| 초대형 (Super-heavy) | 50 t+ | Starship, CZ-9 (계획) |

**이 강좌의 만능 지표: $/kg (kg당 수송단가)**

$$
\text{cost/kg}=\frac{\text{LaunchPrice(PpF)}}{\text{payload}}
$$

## 3. 2025년의 산업 지형

### 3.1 궤도발사 324회+ — 6년 만에 3배

- 2025년 궤도발사: 324~329회(집계기준 상이)
  - 2024년 대비 +25%
  - 2019년 대비 약 3배 — **인류 역사상 최대 발사 활동**
- 성장 동력: **군집위성 배치** — SpaceX 발사의 74%가 자사 Starlink
- 위성 4,500기+ 배치(+54%), 총 질량 ~2,700 t
  - '수요' ✗ / **'자체수요'** ✓ 가 성장을 견인 (11주차 BMC 연계)

> **질문**: 이 성장은 시장의 성장인가, 한 기업의 성장인가?

### 3.2 국가별 지형 — 미·중 2강이 88%

- **미국 193회** — SpaceX가 165회 → '미국의 지배' = '한 기업의 지배'
- **중국 92회** — 10개 조직·25종 기체가 분담 → 국가주도+상업 병행의 이중 트랙 (10주차)
- **유럽 8회·일본 4회** — 전통 우주강국의 체급 대비 왜소한 발사 활동 (10·13주차)
- **한국 2회** — 세계 7위권 진입, 그러나 격차는 '회수'가 아니라 '체제' (14주차)

### 3.3 SpaceX: 한 기업이 세계의 절반을 쏘는 시대

| 지표 | 값 | 해설 |
|---|---|---|
| 2025년 Falcon 9 발사 | **165회** | 세계 나머지 전체보다 많음. 착륙 실패는 단 1회(의도 소모 2회 제외) |
| 궤도 투입 질량 점유율 | **~80%** | 2,213 t (S-1 공시). 2위 중국 CASC의 8배 이상 |
| 자사 발사 중 Starlink 비중 | **74%** | 최대 고객이 자기 자신인 수직 순환 구조 (11주차) |
| 단일 부스터 최다 재비행 | **20회+** | 재사용이 '시연'이 아니라 '운항'이 된 상태 (5주차) |

> **이 강좌의 관점**: 이 지배를 '천재의 신화' ✗ / 학습곡선·발사빈도·자체수요·정부 앵커계약이라는 분석 가능한 메커니즘 ✓ 으로 해부 — 그래야 한국의 대응도 설계 가능

### 3.4 무엇이 이 지형을 만들었나 — 뉴스페이스 4세대 물결

| 세대 | 시기 | 특징 | 내용 |
|---|---|---|---|
| 1세대 | 1980~90s | 상업화의 태동 | Arianespace 설립(1980), 상업 GEO 통신시장 형성 — '정부의 우주'에서 첫 이탈 |
| 2세대 | 2000s | 억만장자 벤처 | SpaceX(2002)·Blue Origin — 실리콘밸리 자본·방식의 진입, COTS 앵커계약(2006) |
| 3세대 | 2010s | VC·스몰샛 붐 | Rocket Lab·Planet 등 수백 개 스타트업, 소형발사체 100여 개 난립(12주차 부침의 씨앗) |
| 4세대 | 2020s | 군집·초대형 | Starlink 메가컨스텔레이션과 Starship — 자체수요·완전재사용이 게임의 규칙을 재정의 |

- **각 세대의 공통 문법**
  - 새 자본(상업계약→벤처→공모시장) 유입 시마다 새 사업모델 등장
  - 정부의 역할: '개발 주체' → '앵커 고객·규제 설계자'로 이동
  - 이 이동의 논리 이해 = 9~11주차의 핵심

**[인터랙티브] 우주 참여기업 지형의 변화 — 매출 규모 네트워크 (2005 ↔ 2025)**

'개발 주체'에서 '앵커 고객·규제 설계자'로 이동해 왔다. 1세대(상업화 태동)에서는 정부가 여전히 직접 개발을 주도했지만, 세대가 진행되며 상업계약→벤처자본→공모시장으로 자본 조달 방식이 바뀌었고 정부는 수요를 보장하는 앵커 고객이자 규제자로 역할을 옮겼다.
</details>

누리호를 통해 발사 횟수 자체는 세계 7위권에 진입했지만, 세계 선도 발사체들은 이미 재사용·메탄 추진·메가컨스텔레이션(군집)이라는 4세대 패러다임으로 이동했다. 즉 한국이 따라잡아야 할 것은 단순 발사 빈도가 아니라 기술·운용 패러다임 자체의 세대 차이다.
</details>
