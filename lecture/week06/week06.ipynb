## PART I · 이론틀 ① 이중용도 배태 — 기술은 태어난 자리를 기억한다

- 기술이 어떤 목적을 위해 만들어졌는가 → 그 기술이 이후 어떤 제도 안에 놓이는가를 결정

> **이론틀 1 · 기술의 사회적 배태(embeddedness)**: 기술은 중립적 도구가 아니라, 그것을 낳은 제도·목적·조직의 흔적을 구조 안에 담는다. 발사체는 (1) 국가가 발주하고, (2) 안보 논리로 정당화되며, (3) 방위산업이 제작하고, (4) 무기 통제 체제 안에서 관리되는 형태로 태어났다. 이후 상업화가 진행되어도 이 네 겹의 제도적 껍질은 그대로 남아, 신규 진입자마저 그 틀 안에서 활동하게 된다.

| 제도적 껍질 | 기원 | 오늘날 어디에서 나타나는가 |
|---|---|---|
| 국가 발주 | 초기 수요가 전적으로 정부 | 지금도 다수 국가에서 정부가 최대 고객 — 12주차 |

| 방산 제작 | 제작사가 방위산업체 | 원가보상형 계약 관행과 비용 구조가 여기서 유래 — 11주차 |
| 무기 통제 | 기술·부품이 통제 목록에 포함 | MTCR·수출통제·발사 허가 제도의 근거 — 7주차 |

**반론도 함께 볼 것**
- 배태 논리를 과도하게 적용하면 '기술이 운명을 결정한다'는 결정론에 빠짐
- 실제로는 같은 기원에서 출발한 국가들이 서로 다른 제도로 갈라짐 — Part IV의 비교가 그 증거

## PART II · 이론틀 ② 값비싼 신호 — 왜 국가는 상징에 돈을 쓰는가

- "국가 위신"은 흔히 비합리적 낭비로 취급 — 그러나 신호이론으로 보면 상당히 합리적인 행동일 수 있음

> **이론틀 2 · 값비싼 신호(costly signaling)**: 국가는 자신의 능력과 의지를 상대에게 알리고 싶지만, 말은 값이 들지 않으므로 믿을 수 없다. 신뢰할 수 있는 신호가 되려면 (1) 능력이 없는 쪽은 흉내 낼 수 없을 만큼 비싸고, (2) 성공·실패가 공개적으로 검증 가능해야 한다. 우주발사는 이 두 조건을 거의 완벽하게 만족한다.

| 신호로서의 조건 | 우주발사가 이를 만족하는 방식 | 충족도 |
|---|---|---|
| 모방 불가능성 | 장기간의 자본·인력·산업 축적 없이는 성공할 수 없음 | 높음 |
| 공개 검증 가능성 | 궤도 물체는 제3자가 독립적으로 추적·확인 가능 | 매우 높음 |
| 실패의 가시성 | 실패도 똑같이 공개됨 — 신호의 신뢰도를 높이는 요소 | 높음 |
| 군사 역량과의 상관 | 발사 능력은 미사일 능력과 강하게 연결 | 직접적 |

**그래서 관찰되는 현상들**
- 경제 규모에 비해 과도해 보이는 우주 예산을 편성하는 국가가 반복적으로 나타남
- "최초" 타이틀에 자원이 집중 — 두 번째 성취는 신호 가치가 급락하기 때문
- 자국 발사체 보유 자체가 목표가 되고, 해외 발사 구매는 비용이 싸도 선호되지 않음
- 유인 우주비행처럼 비용 대비 과학적 산출이 낮은 사업이 정치적으로 강한 지지를 받음
- 반대로 신호가 필요 없어지면(냉전 종식) 예산이 급감

> **주의**: 이 틀은 위신 추구를 '설명'하지만 '정당화'하지는 않는다. 신호 가치가 실제 사회적 편익을 초과하는지는 별도의 비용편익 판단이 필요하며, 이는 Part V의 공공선택론적 비판으로 이어진다.

## PART III · 설계는 어떻게 결정되었는가 — 세 개의 압력

- 기술자들이 최선이라 생각한 형상은 따로 있었음 — 최종 형상은 세 가지 정치·예산 압력이 교차한 지점에서 결정

1. **예산 상한** — 완전재사용 2단 형상은 초기 개발비가 지나치게 컸음. 예산 당국은 개발비 상한을 먼저 정함. → 결과: 부분재사용으로 후퇴(외부 탱크는 버리고, 부스터는 회수)
2. **군의 요구조건** — 군을 고객으로 확보해야 예산 정당화가 가능했음. 그 대가로 대형 화물창과 큰 교차범위(귀환 시 옆으로 이동하는 능력)가 요구조건에 포함. → 결과: 기체가 커지고 날개가 커짐(질량과 복잡도 증가)
3. **수요의 독점화** — 여러 수송수단을 유지할 예산이 없어 모든 임무를 하나로 처리한다는 전제가 세워짐. → 결과: 소모성 발사체 생산이 축소되고 대안이 사라짐

> **고체 부스터 선택 — 정치지리가 기술을 결정한 사례**: 고체 부스터는 액체 부스터보다 초기 개발비가 낮다는 이유로 선택되었다. 그런데 제작지에서 발사장까지 운반해야 했기 때문에 부스터를 여러 조각으로 나눠 만들고 현장에서 접합하는 구조가 되었다. 1986년 사고의 직접 원인이 된 것이 바로 이 접합부의 밀봉이었다. 예산 제약이 형상을 정하고, 산업 입지가 구조를 정하고, 그 구조가 사고 지점이 되었다.

## PART III · 이론틀 ③ 예산제약형 설계의 함정

- 개발비를 줄이라는 요구는 합리적으로 보이나, 그 요구가 생애주기 비용을 오히려 키우는 경우가 체계적으로 존재

> **이론틀 3 · Design-to-budget과 비용의 이전(cost shifting)**: 예산 심의는 대개 연도별 지출을 통제한다. 그 결과 설계자는 '총 생애주기 비용 최소화'가 아니라 '초기 개발비 상한 준수'를 목적함수로 삼게 된다. 두 목적함수의 해는 다르다. 초기 비용을 줄이는 선택(부분재사용, 고체 부스터, 검증 축소)은 대개 운용 단계의 비용과 위험을 키운다. 즉 비용은 사라진 것이 아니라 미래로, 그리고 다른 회계 항목으로 이전된 것이다.

| 선택 | 단기 효과 | 이전된 비용 |
|---|---|---|
| 완전재사용 → 부분재사용 | 개발비 절감 | 매 비행마다 탱크 제작·부스터 회수 정비 비용 발생 |
| 액체 부스터 → 고체 부스터 | 개발비 절감 | 정비 부담과 안전 위험 증가 |
| 시제기 제작 축소 | 개발비 절감 | 설계 여유 부족이 운용 제약으로 나타남 |
| 군 요구조건 수용 | 예산 지지 확보 | 기체 대형화 → 제작·운용비 상승 |

**일반화 — 어디서든 반복되는 구조**
- 예산 주기가 짧고 감사 기준이 연도별 지출이면, 생애주기 최적화는 제도적으로 불리
- 개발 조직과 운용 조직이 분리되어 있으면 비용 이전을 아무도 막지 않음
- 해법은 기술이 아니라 회계·평가 제도 — 생애주기 비용을 심의 기준에 포함시키는 것
- 국내에서도 예비타당성조사와 총사업비 관리 방식이 같은 문제를 안고 있는지 점검 필요

> **연결**: 4주차에서 본 '단번 성공형 vs 반복 시험형' 선택도 결국 같은 문제다. 제도가 무엇을 비용으로 계상하는가가 기술적 선택을 만든다.

## PART III · 단일 수송수단 전략과 그 붕괴

- 모든 발사를 하나로 처리한다는 결정은 효율적으로 보였음 — 그러나 시스템 전체를 하나의 고장점에 묶는 결정이기도 했음

1. **수요 통합** — 모든 정부·상업·군 탑재체를 셔틀로 일원화하는 방침
2. **대안 축소** — 소모성 발사체 생산라인이 축소되고 대안이 사라짐
3. **사고와 정지** — 1986년 사고로 비행이 장기간 중단, 모든 발사가 함께 멈춤
4. **정책 전환** — 상업 탑재체를 셔틀에서 배제하고 소모성 발사체를 되살림
5. **제도 형성** — 정부 수요를 민간 발사체로 조달하는 체계가 자리 잡음

**무엇이 문제였나 — 회복력의 관점**
- 단일 수단 전략은 평시 효율은 높지만 단일 고장점(single point of failure)을 만듦
- 사고 한 번으로 국가의 모든 우주 접근이 중단 — 안보상 받아들이기 어려운 취약성
- 그 경험이 '보장된 우주 접근(assured access)' 원칙을 낳음 — 이후 복수 발사체 유지 정책의 근거(9주차)

**역설적 결과 — 실패가 산업을 만들었다**
- 상업 탑재체가 셔틀에서 밀려나면서 민간 발사서비스 시장이 열림
- 정부가 발사체를 직접 운영하는 대신 발사 '서비스'를 구매하는 조달 방식이 자리 잡음
- 이 전환이 없었다면 오늘날의 상업 발사 산업은 성립하기 어려웠을 것 — 정책 실패가 의도치 않게 제도 혁신을 촉발한 사례(9주차 미국 정책의 출발점)

## PART III · 다섯 가지 정책 교훈

- 각 교훈은 오늘의 의사결정에 그대로 적용할 수 있는 형태로 정리

1. **약속된 발사 빈도를 검증하라** — 재사용의 경제성은 비행 횟수에 전적으로 의존. 사업 계획서의 발사 빈도 가정이 수요 분석에 근거하는지, 아니면 비용을 맞추기 위해 역산된 숫자인지 확인 필요
2. **요구조건의 출처를 추적하라** — 성능 요구조건 중 일부는 임무가 아니라 예산 연합을 만들기 위해 들어옴. 그런 요구조건은 비용을 키우면서 실제로는 거의 사용되지 않음
3. **생애주기 비용으로 심의하라** — 개발비 상한만 통제하면 비용은 운용 단계로 이전될 뿐. 심의 기준 자체가 설계를 결정
4. **단일 수단에 수요를 몰지 마라** — 효율과 회복력은 상충. 안보상 중요한 기능일수록 중복은 낭비가 아니라 보험
5. **실패의 정치적 비용을 미리 설계하라** — 사고 후 대응 방식이 정해져 있지 않으면 프로그램 전체가 정지. 사고 조사·복귀 절차·대체 수단이 사전에 제도화되어야 함

## PART III · 왜 셔틀은 30년이나 유지되었는가

- 문제가 일찍 드러났는데도 대체되지 않음 — 이 지속성 자체가 분석 대상, Part V의 경로의존성으로 넘어가는 다리

- **대체재 부재**: 유인 우주비행 수단이 셔틀뿐이었고, 우주정거장 조립이 셔틀 사양에 맞춰 설계되어 있었음
- **매몰비용 논리**: 이미 투입한 개발비를 정당화하려면 계속 운영해야 한다는 압력이 작동
- **고용과 지역 이해**: 수만 명 규모의 고용과 여러 지역에 걸친 산업 기반이 형성되어 있었음
- **대체 계획의 실패**: 후속 재사용 발사체 개발 시도들이 기술·예산 문제로 잇따라 중단

> **이론틀 3b · 매몰비용 오류와 정책의 지속 — 왜 나쁜 사업은 끝나지 않는가**: 경제학적으로 매몰비용은 의사결정에 반영해서는 안 된다. 그러나 정치 과정에서는 반대다. 사업을 중단하면 '그동안의 예산이 낭비되었다'는 평가가 따르고, 그 평가의 대가를 특정 정치 주체가 부담한다. 반면 사업을 계속하면 비용은 미래로 분산되고 책임 주체는 흐려진다. 그래서 매몰비용이 클수록 중단이 어려워진다 — 합리적 개인들의 합리적 선택이 집합적으로 비합리적 결과를 낳는 전형적 구조다.

## PART IV · 소련 — 설계국 경쟁 체제의 힘과 한계

- 초기의 압도적 성취와 달 경쟁의 좌절이 같은 제도에서 나옴 — 제도의 구조를 보면 두 결과가 모순이 아님을 알 수 있음

- 제도의 구조: 복수의 설계국이 각각 총설계자를 중심으로 독립적으로 경쟁, 자체 시제 제작 능력을 갖추고 상부에 직접 제안. 최고 지도부가 사업을 직접 선택하며 중간 조정 기구가 약하나, 군수 체계 안에 편입되어 있어 자원 동원력은 매우 강함

| 국면 | 제도의 작용 | 구체적 양상 |
|---|---|---|
| 초기 성취 | 강점으로 작용 | 빠른 의사결정과 자원 집중, 기존 미사일의 신속한 전용으로 연속적 '최초'를 달성 |
| 달 착륙 경쟁 | 약점으로 작용 | 설계국 간 경쟁이 중복 개발과 자원 분산을 낳고, 대형 발사체 개발에 통합 조정이 이뤄지지 않음 |
| 기술 계보 | 독자성 확보 | 산화제과잉 다단연소 등 서방과 다른 계보를 형성(3주차) — 후일 국제 이전의 자산이 됨 |
| 체제 전환 이후 | 취약성 노출 | 수요 급감과 인력 유출, 생산 기반의 국가 분리로 회복이 지연 |

**일반화할 수 있는 교훈**
- 경쟁은 초기 탐색 단계에서 유효, 대규모 통합 개발 단계에서는 조정이 더 중요 — 사업 단계에 따라 최적의 거버넌스가 달라짐
- '누가 최종 통합 책임을 지는가'가 명확하지 않은 체제는 규모가 커질수록 실패 확률이 높아짐 — 국내 체계종합 주관기관 논의에도 그대로 적용

> **주의**: 소련의 좌절을 '계획경제의 실패'로 단순화하는 서술은 부정확하다. 같은 체제가 초기에는 앞섰다는 사실이 그 설명을 반박한다.

## PART IV · 유럽 — 실패한 첫 시도가 성공한 두 번째를 설계했다

- 유럽의 발사체 협력은 처음에 완전히 실패 — 그 실패의 원인 분석이 지금까지 이어지는 산업정책 원칙을 낳음

- **1차 시도 — 다국적 분담 개발**: 각국이 단(stage)을 하나씩 맡아 개발하고 조립하는 방식. 궤도 투입에 끝내 성공하지 못하고 사업이 종료. 실패 원인: 시스템 통합의 최종 책임 주체 부재, 인터페이스 관리 실패, 회원국 간 일정·품질 편차
- **2차 시도 — 단일 주관 + 산업 배분**: 하나의 기관이 사업 전체를 주관하고 통합 책임을 지되, 산업 계약은 회원국 기여 비율에 따라 배분하는 방식으로 전환. 결과: 상업 발사서비스 시장에서 장기간 주도적 지위 확보

> **이론틀 4a · 지리적 환류(geo-return) 원칙**: 회원국이 낸 분담금에 비례해 자국 기업에 산업 계약이 돌아가도록 배분하는 원칙. 이것이 없으면 각국이 분담금을 낼 정치적 이유가 사라져 협력 자체가 성립하지 않는다. 즉 geo-return은 비효율이 아니라 협력을 가능하게 한 대가다. 문제는 그 대가가 이후 경쟁 환경에서 비용 구조의 경직성으로 나타난다는 점이다.

**오늘의 딜레마**
- 저가 경쟁이 심화되자 산업 배분 원칙이 비용 절감의 걸림돌로 지목되기 시작
- 원칙을 폐기하면 회원국의 참여 유인이 약해짐 — 협력 구조 자체가 흔들림
- '효율을 위해 정치적 기반을 허물 수 있는가'라는 전형적 정책 딜레마
- 10주차에서 유럽의 대응(경쟁 도입 실험)을 상세히 다룸

## PART IV · 일본 — 기술 도입에서 자립으로, 그 전환의 대가

- 가장 빠른 길로 시작해 가장 어려운 길로 갈아탄 사례 — 후발국에게 가장 직접적인 시사점

| 단계 | 전략 | 얻은 것 | 치른 대가 |
|---|---|---|---|
| 1단계 — 기술 도입 | 해외 발사체 기술을 라이선스로 도입해 국산화율을 점진적으로 높임 | 빠른 능력 확보 | 핵심 설계 역량은 이전되지 않음·사용·수출에 제약 |
| 2단계 — 자립 개발 | 대형 수소 엔진을 포함해 전 계통을 독자 개발 | 설계 주권 확보 | 개발 비용과 기간이 크게 증가·초기 신뢰성 문제 |
| 3단계 — 저비용화 | 자립 형상을 유지하되 비용 구조를 재설계 | 상업 경쟁력 시도 | 이미 굳어진 산업 구조 안에서의 절감에는 한계 |
| 4단계 — 세대 교체 | 차세대 기체로 구조적 비용 절감을 시도 | 진행 중 | 초기 비행 실패를 겪으며 신뢰성 회복 과제 |

**후발국에 주는 교훈**
- 기술 도입은 시간을 사지만 역량을 사지는 못함(Part I의 암묵지 논의)
- 자립 전환은 어느 시점에 하든 비용이 크며, 늦출수록 도입 기술에 대한 의존이 깊어짐
- 자립 이후에는 '비용 경쟁력'이라는 완전히 다른 과제가 기다림 — 즉 자립은 종착점이 아니라 두 번째 출발선

**한국과의 대비**
- 한국은 초기 협력(나로호) 이후 비교적 이른 시점에 자립 개발(누리호)로 전환
- 따라서 지금 서 있는 지점은 일본의 3단계 — '자립은 했으나 비용 경쟁력은 이제부터'에 해당
- 차세대발사체의 재사용·메탄 전환은 이 두 번째 과제에 대한 답으로 읽을 수 있음(3주차)

## PART IV · 같은 기원, 다른 제도 — 무엇이 분기를 만들었나

- 네 사례 모두 군사 기술에서 출발 — 그런데도 서로 다른 제도로 갈라졌다면, 기원은 결과를 확정하지 않는다는 뜻

| 국가·지역 | 출발점 | 초기 제도 선택 | 경로의 굴곡 | 현재의 구조 |
|---|---|---|---|---|
| 미국 | 군 미사일 계열 + 인력 흡수 | 민간 기관 창설로 민군 분리 | 단일 수단 집중 → 실패 → 서비스 구매 전환 | 정부가 고객, 민간이 운영 |
| 소련·러시아 | 군 미사일 계열 | 군수 체계 내 설계국 경쟁 | 통합 조정 취약 → 대형 사업 좌절 | 국가 통합 기업 체제 |
| 유럽 | 각국 개별 개발 | 다국적 협력 기구 + 산업 배분 | 1차 실패 → 단일 주관으로 재설계 | 협력 유지와 비용 경쟁의 긴장 |
| 일본 | 해외 기술 도입 | 민간 기관 + 산업체 중심 | 자립 전환 후 비용 재설계 반복 | 자립 유지와 상업성 확보의 병행 |

**분기를 만든 변수**
- 누가 통합 책임을 지는가 (단일 주관 vs 분산)
- 수요가 하나인가 여럿인가 (독점 vs 복수 수단)
- 산업 배분이 정치적 전제인가 아닌가
- 자립을 언제, 어느 범위까지 추구했는가

> **분석적 결론**: 군사적 기원은 모든 국가에 공통된 초기 조건이었지만, 이후의 제도 선택이 결과를 갈랐다. 따라서 정책 분석의 초점은 '기원'이 아니라 '기원 이후의 선택'에 놓여야 한다. 그리고 그 선택은 대개 되돌리기 어렵다 — 다음 파트의 주제다.

## PART V · 이론틀 ④ 경로의존성 — 왜 되돌릴 수 없는가

- 초기의 작은 선택이 이후의 선택지를 체계적으로 좁히는 현상 — 우주수송 정책은 이 개념의 교과서적 사례를 여럿 제공

> **이론틀 4 · 경로의존성(path dependence)과 고착(lock-in)**: 어떤 기술·제도가 우연한 초기 조건으로 채택되면, 이후 (1) 학습효과, (2) 조정효과, (3) 적응적 기대, (4) 규모의 경제가 작동해 그 선택을 강화한다. 그 결과 더 나은 대안이 나타나도 전환 비용이 이익을 초과해 기존 경로가 유지된다. 중요한 것은 이것이 비합리적 행동이 아니라, 개별 주체에게는 합리적인 선택의 집합적 결과라는 점이다.

| 고착을 만드는 메커니즘 | 작동 방식 | 우주수송 분야의 사례 |
|---|---|---|
| 학습 효과 | 특정 기술에 숙련이 쌓일수록 다른 기술의 상대적 비용이 커짐 | 케로신 인프라·인력이 있는 조직에서 메탄 전환이 어려운 이유(3주차) |
| 조정 효과 | 주변 시스템이 그 기술을 전제로 설계됨 | 우주정거장 조립이 특정 발사체 사양에 맞춰진 사례(Part III) |
| 적응적 기대 | 다들 그 기술이 계속될 것이라 믿고 투자 | 공급망·부품업체가 특정 계열에 특화 |
| 규모의 경제 | 누적 생산이 단가를 낮춰 후발 대안이 불리해짐 | 학습곡선이 진입장벽으로 작용(3·12주차) |

**정책적 함의 — 초기 결정의 무게**
- 발사장 입지, 추진제 선택, 사업 주관 구조 같은 초기 결정은 수십 년을 구속(4주차 발사장 논의)
- 초기 결정에는 그 시점의 최적성뿐 아니라 미래의 선택지를 얼마나 남기는가(옵션 가치)를 함께 따져야 함
- 반대로 이미 고착된 것을 바꾸려면 기술적 우위만으로는 부족, 전환 비용을 흡수할 별도의 정책 수단이 필요

## PART V · 이론틀 ⑤ 공공선택론 — 누구의 이익이 제도를 지탱하는가

- 정책을 '국가 이익의 추구'로만 보면 설명되지 않는 현상이 많음 — 정책 결정자와 이해집단의 유인을 함께 봐야 함

- **분산된 편익 구조**: 대형 사업의 계약과 고용이 여러 지역에 나뉘어 배치되면, 각 지역의 대표자가 사업 존속을 지지할 유인을 갖게 됨 → 사업 축소·종료가 정치적으로 매우 어려워짐
- **집중된 이익 vs 분산된 비용**: 이익은 소수 산업·지역에 집중되고 비용은 전체 납세자에게 분산 — 이익 집단은 조직되지만 비용 부담자는 조직되지 않음 → 비용편익이 불리해도 사업이 유지됨
- **관료 조직의 유인**: 기관은 예산과 인력 규모를 유지·확대하려는 유인을 가짐 — 사업 종료는 조직 축소를 뜻함 → 대체 사업이 없으면 기존 사업 유지를 선호
- **정보 비대칭**: 사업의 기술적 타당성을 평가할 역량이 수행 기관에 집중 → 외부 감시가 형식화되기 쉬움

> **균형 잡힌 사용법**: 공공선택론은 강력한 설명 도구지만, 모든 정책을 '이익집단의 포획'으로 환원하면 그 자체가 또 다른 단순화가 된다. 실제 정책에는 진지한 전략적 판단과 이해관계가 함께 작동한다. 분석의 목표는 어느 쪽이 진짜인지 고르는 것이 아니라, 두 요소가 각각 어느 정도의 비중으로 작동하는지 가려내는 것이다. 10주차와 14주차에서 이 도구를 실제 사례에 적용한다.

## PART V · 이론틀 ⑥ 시장형성의 경제학 — 왜 지금 시장으로 이행하는가

- 방금 본 '무엇이 단절되었는가'(수요의 원천·조달 방식)는 현상 기술이다 — 하버드 경영대학원 매슈 와인지얼(Weinzierl, 2018)은 그 현상이 **왜 지금 가능해졌는지**를 경제학적으로 설명한다

> **이론틀 6 · 시장형성(market formation)의 조건**
> Weinzierl, "Space, the Final Economic Frontier," *JEP* (2018); Weinzierl & Rosseau, *Space to Grow* (2025)
>
> **시장은 구매자의 지불의사(WTP)가 생산비용 + 규제 불확실성의 합을 초과할 때 형성된다.**
>
> WTP > 생산비용(C) + 규제 리스크 비용(R) — 이 부등식이 성립하는 순간, 정부가 유일한 구매자·공급자였던 시장에 민간 기업이 뛰어들 유인이 생긴다.

**이 부등식을 충족시키는 세 축 — 6주차에서 본 '제도적 껍질'이 왜, 어떻게 벗겨지고 있는가**

| 축 | 과거 (0단계 · 정부독점) | 현재 (1단계 · 시장형성) |
|---|---|---|
| ① 공급 — 비용 | 방산 원가보상형 계약(11주) → 경쟁 유인 부재, 비용 절감 무의미 | 재사용·학습곡선(5·12주) → SpaceX發 공급충격, 진입비용 급락 |
| ② 자본 — 투자 | 국가 발주 독점(Part I) → 민간 자본이 들어올 시장 자체가 부재 | 뉴스페이스 VC·공모자금 → 연 5억$(2008)→25억$(2016)대 성장(1주 4세대 물결) |
| ③ 제도 — 규제 | 안보 정당화·무기통제(Part I·7주 예고) → 규제 불확실성이 WTP를 상쇄 | COTS 앵커계약·발사허가 간소화(9주) → 규제자가 시장을 '설계'하는 레버로 전환 |

> **정리** — 시장은 저절로 생기지 않는다. 기술(비용)·자본(투자)·제도(규제설계) 세 축이 동시에 임계점을 넘어야 '0단계 정부독점'이 '1단계 시장형성'으로 이행한다.


**[인터랙티브] 시장형성 탐색기** — 아래에서 지불의사(WTP)·생산비용·규제 리스크비용을 직접 조작해, 어느 축을 움직여야 부등식(WTP > 비용+규제리스크)이 성립하는지 확인한다. 정부가 공급·자본·제도 중 무엇을 손대야 하는지에 대한 직관을 만드는 것이 목적이다.

```python
def _plot_market_formation(wtp_musd, cost_musd, reg_risk_musd):
    total_barrier = cost_musd + reg_risk_musd
    gap = wtp_musd - total_barrier
    forms = gap > 0
    bar_color = GREEN if forms else RED

    fig, ax = plt.subplots(figsize=(8.5, 5))
    ax.set_facecolor("white")
    for s in ["top", "right"]:
        ax.spines[s].set_visible(False)
    for s in ["left", "bottom"]:
        ax.spines[s].set_color("#B0B0B0")
    ax.grid(True, color=GRID, linewidth=0.8, zorder=0, axis="y")
    ax.tick_params(colors=INK_MUTED, labelsize=10)

    ax.bar([0], [wtp_musd], color=PRIMARY, width=0.5, zorder=3, label="지불의사 WTP")
    ax.bar([1], [cost_musd], color=AMBER, width=0.5, zorder=3, label="생산비용")
    ax.bar([1], [reg_risk_musd], bottom=[cost_musd], color=RED, width=0.5, zorder=3, label="규제 리스크비용")

    ax.set_xticks([0, 1])
    ax.set_xticklabels(["지불의사(WTP)", "비용 + 규제리스크"])
    ax.set_ylabel("$M / 발사 (가정)")
    ax.legend(loc="upper right", fontsize=9, frameon=False)

    verdict = "시장형성 조건 충족 → 민간기업 진입 유인 발생" if forms else "시장형성 조건 미충족 → 정부 개입 없이는 형성 어려움"
    ax.set_title(f"WTP ${wtp_musd:.0f}M vs 비용+리스크 ${total_barrier:.0f}M  →  격차 ${gap:+.0f}M  ({verdict})",
                 fontsize=11, color=bar_color)
    plt.tight_layout()
    plt.show()

    print("[해석] Weinzierl(2018) 모형: 시장은 WTP > 비용+규제불확실성일 때 형성됩니다.")
    print(f"  현재 가정 — WTP ${wtp_musd:.0f}M · 생산비용 ${cost_musd:.0f}M · 규제리스크비용 ${reg_risk_musd:.0f}M")
    if forms:
        print(f"  → 격차 +${gap:.0f}M: 민간 기업이 이 시장에 진입할 경제적 유인이 존재합니다.")
    else:
        print(f"  → 격차 ${gap:.0f}M(부족): 정부가 공급(비용↓)·자본(투자유치)·제도(규제레버) 중 하나 이상을 움직이지 않으면 시장이 형성되지 않습니다.")


def market_formation_explorer():
    """와인지얼의 시장형성 부등식(WTP > 비용+규제불확실성)을 슬라이더로 조작 —
    공급(비용)·자본·제도(규제리스크) 축 중 어디를 움직여야 시장이 형성되는지 직접 확인한다."""
    interact(_plot_market_formation,
             wtp_musd=FloatSlider(value=45, min=5, max=100, step=5, description="지불의사 WTP($M)",
                                   style={"description_width": "140px"}, layout={"width": "480px"}),
             cost_musd=FloatSlider(value=30, min=5, max=100, step=5, description="생산비용($M)",
                                    style={"description_width": "140px"}, layout={"width": "480px"}),
             reg_risk_musd=FloatSlider(value=10, min=0, max=60, step=5, description="규제 리스크비용($M)",
                                        style={"description_width": "140px"}, layout={"width": "480px"}))

market_formation_explorer()
```

## PART V · 이론틀 ⑥ 확장 — 왜 보조금인가: Chicken-and-Egg 문제와 빅푸시(big push)

- 앞의 부등식(WTP > C+R)은 **정태적(static)** 조건이다 — 어느 한 시점에서 시장이 성립하는지만 판정함
- 그러나 발사체의 비용(C)은 고정값이 아니라 **누적 생산량의 함수**다(3·5·12주차 학습곡선) — 수요가 있어야 생산량이 늘고, 생산량이 늘어야 비용이 떨어지고, 비용이 떨어져야 수요가 붙는 순환 구조
- 이것이 **치킨앤에그(chicken-and-egg) 문제**다 — 와인지얼은 이 순환을 끊지 못하면 시장은 낮은 물량에서 정체된 채 스스로 성장 궤도에 올라서지 못한다고 봄
- 우주 분야는 이 문제가 특히 심각한데, 개발→양산→학습곡선 하강까지 걸리는 시간이 길어 **인내자본(patient capital)**이 대규모로 필요하기 때문 — 민간 자본은 그 '계곡(valley)'을 버틸 유인이 약함

> **빅푸시(big push) 이론** (Rosenstein-Rodan, 1943의 논리를 시장형성에 적용)
>
> 낮은 물량의 '정체 균형'과 높은 물량의 '자력성장 균형' 사이에는 불안정한 임계 물량(Q\*)이 존재한다. 이 임계점 아래에서는 아무리 기다려도 시장이 저절로 자라지 않는다 — 누군가 한 번에 임계점을 넘겨야(big push) 학습곡선이 스스로 WTP 아래로 비용을 밀어내는 자기강화 구간에 진입한다.

**보조금의 역할 — 정부는 두 가지 방식으로 '한 번의 밀어올림'을 제공할 수 있음**

| 방식 | 메커니즘 | 사례 |
|---|---|---|
| 공급측 보조금 | 개발비·생산비 직접 지원 → 비용곡선 C(Q) 자체를 아래로 이동 | 발사체 개발 국가R&D, 세액공제 |
| 수요측 보조금(앵커계약) | 정부가 물량을 선구매·보증 → 물량(Q)을 인위적으로 임계점 너머로 이동 | NASA COTS/CRS 앵커테넌시 |

> **핵심 통찰** — 보조금의 목표는 "영원히 지원하는 것"이 아니라 "시장이 스스로 자라는 구간까지만 밀어주는 것"이다. 임계점을 넘기고 나면 학습곡선이 알아서 비용을 낮추므로 보조금은 설계상 **시한**을 가져야 한다 — 그렇지 않으면 Part III에서 본 매몰비용·경로의존 함정에 다시 빠진다.


**[인터랙티브] 빅푸시 탐색기** — 학습률(학습곡선 기울기), 목표 WTP, 정부가 보조하는 초기 물량(Q0)을 조작해 시장이 임계 물량(Q*)을 넘어 자력성장 구간에 들어서는지 확인한다. 빨간 구간은 '정체(chicken-and-egg)', 초록 구간은 '자력성장'이다.

```python
def _plot_subsidy_bigpush(learning_rate_pct, wtp_musd, subsidized_q0):
    # 가정: 파일럿(1호기) 발사 비용 C0 = $150M — 개념 설명용 근사치
    C0 = 150.0
    b = -np.log2(1 - learning_rate_pct / 100.0)   # Wright's law 학습곡선 지수
    Q = np.linspace(1, 300, 600)
    C = C0 * Q ** (-b)
    Q_star = (C0 / wtp_musd) ** (1 / b)            # 비용곡선과 WTP가 만나는 임계 물량
    escaped = subsidized_q0 >= Q_star

    fig, ax = plt.subplots(figsize=(8.5, 5))
    ax.set_facecolor("white")
    for s in ["top", "right"]:
        ax.spines[s].set_visible(False)
    for s in ["left", "bottom"]:
        ax.spines[s].set_color("#B0B0B0")
    ax.tick_params(colors=INK_MUTED, labelsize=10)

    q_star_clip = min(Q_star, 300)
    ax.axvspan(1, q_star_clip, color=RED, alpha=0.08, zorder=0)
    ax.axvspan(q_star_clip, 300, color=GREEN, alpha=0.08, zorder=0)

    ax.plot(Q, C, color=PRIMARY, linewidth=2.4, zorder=3, label="비용곡선 C(Q) — 학습곡선")
    ax.axhline(wtp_musd, color=AMBER, linestyle="--", linewidth=1.6, zorder=3, label="지불의사 WTP")

    if Q_star <= 300:
        ax.plot(Q_star, wtp_musd, "o", color=INK, markersize=9, zorder=5)
        ax.annotate(f"임계물량 Q*≈{Q_star:.0f}", (Q_star, wtp_musd), fontsize=9, color=INK,
                    xytext=(8, 10), textcoords="offset points")

    q0_color = GREEN if escaped else RED
    ax.axvline(subsidized_q0, color=q0_color, linestyle=":", linewidth=2, zorder=4)
    ax.annotate(f"보조금 물량 Q0={subsidized_q0}", (subsidized_q0, ax.get_ylim()[1] * 0.92),
                fontsize=9, color=q0_color, xytext=(6, 0), textcoords="offset points")

    ax.set_xlabel("누적 생산·발사 물량 Q (기)")
    ax.set_ylabel("$M / 기")
    ax.legend(loc="upper right", fontsize=9, frameon=False)

    verdict = "탈출 성공 — 자력성장 구간 진입" if escaped else "탈출 실패 — 여전히 정체 구간"
    ax.set_title(f"학습률 {learning_rate_pct}% · WTP ${wtp_musd:.0f}M · Q0={subsidized_q0}  →  {verdict}",
                 fontsize=11, color=q0_color)
    plt.tight_layout()
    plt.show()

    print(f"[해석] 임계 물량 Q* ≈ {Q_star:.0f}기 — 이 물량을 넘어야 비용이 WTP 아래로 내려가 시장이 스스로 굴러갑니다.")
    if escaped:
        print(f"  → 보조금으로 확보한 물량 Q0={subsidized_q0}기가 Q*를 넘어섰습니다: 이후로는 보조금 없이도 학습곡선이 비용을 계속 낮춥니다.")
        print("  → 정책 함의: 이 시점 이후의 보조금 연장은 효율성 없는 지대추구가 될 위험이 커집니다 — 출구 설계가 필요합니다.")
    else:
        print(f"  → 보조금 물량이 부족합니다: Q0={subsidized_q0}기 < Q*={Q_star:.0f}기. 여전히 chicken-and-egg 정체 구간입니다.")
        print("  → 정책 함의: 물량을 늘리거나(수요측 보조금 확대) 비용곡선 자체를 낮춰야(공급측 보조금) 임계점을 넘길 수 있습니다.")
    print("  ※ C0=$150M(가정), 학습곡선·WTP 수치는 개념 설명용 근사치이며 실제 프로그램 원가와 다를 수 있습니다.")


def subsidy_bigpush_explorer():
    """학습곡선(치킨앤에그) 상에서 보조금이 만드는 '빅푸시'를 슬라이더로 조작 —
    보조금 물량(Q0)이 임계 물량(Q*)을 넘어서야 시장이 자력성장 구간에 진입함을 확인한다."""
    interact(_plot_subsidy_bigpush,
             learning_rate_pct=IntSlider(value=15, min=8, max=25, step=1, description="학습률(%)",
                                          style={"description_width": "140px"}, layout={"width": "480px"}),
             wtp_musd=FloatSlider(value=60, min=20, max=150, step=10, description="지불의사 WTP($M)",
                                   style={"description_width": "140px"}, layout={"width": "480px"}),
             subsidized_q0=IntSlider(value=20, min=1, max=150, step=1, description="보조금 물량 Q0(기)",
                                      style={"description_width": "140px"}, layout={"width": "480px"}))

subsidy_bigpush_explorer()
```

> **보고서 연계** — 「③ 시장분석」 파트를 쓸 때, 팀은 먼저 이 표의 0~3단계 중 자신들이 가정하는 목표 시장이 몇 단계에 있는지 선언해야 한다. 1단계(형성)를 가정하면서 2단계(정교화) 도구인 가격 벤치마크를 근거 없이 들이대는 것은 논리적 모순이다.

- 개발비 상한 중심의 심의가 비용을 운용 단계로 이전시킨다면, 총사업비 관리 제도를 어떻게 바꿔야 하는가. 그 변경이 만들 새로운 부작용은 무엇인가

**Q5. 규제 불확실성과 위신 논리의 상충**

**Q6. 한국의 시장형성 조건**
- 한국의 발사서비스 수요만으로 WTP > 비용+규제리스크 부등식이 성립할 수 있는가. 성립하지 않는다면 정부는 공급·자본·제도 중 어느 축을 먼저 움직여야 하는가

**Q7. 정교화 도구의 시기**
- '시장정교화'(2단계) 도구—보험, 표준화, 가격벤치마크—를 시장형성(1단계)조차 안 된 국내 시장에 도입하는 것은 시기상조인가, 아니면 형성을 앞당기는 촉매인가

**Q8. 보조금의 출구 설계**
- 임계 물량(Q*)을 넘긴 이후에도 보조금이 계속된다면 어떤 정치·제도적 유인이 그 지속을 만드는가(Part V 공공선택론과 연결). 보조금에 '자동 일몰' 조항을 설계한다면 그 기준(물량인가, 시간인가, 가격인가)은 무엇이어야 하는가
