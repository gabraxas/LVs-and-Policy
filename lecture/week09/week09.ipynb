## PART 1 · 출발점 — 국가독점 시대와 그 균열

**1958 — 국가항공우주법**
- NASA 창설. 냉전기 국가 위신·안보 경쟁의 도구로서 우주개발이 전적으로 연방정부 사업으로 출발 — 발사체는 국가자산, 민간의 역할은 하도급 제작에 국한

**1981~ — 우주왕복선의 '독점 운송인' 지위**
- NASA는 우주왕복선을 상업위성 발사까지 포괄하는 국가 표준 수송수단으로 정책화 — 민간 소모성발사체(ELV) 시장은 사실상 봉쇄

**균열의 신호**
- 왕복선의 발사 지연·높은 운영비가 누적되며, 위성사업자들 사이에서 '민간 대체 수송수단이 필요하다'는 요구가 산업계로부터 제기되기 시작

## PART 1 · 전환점 ① — 1984년 상업우주발사법(CSLA)

1984년, 미국은 세계 최초로 민간 발사체 사업을 합법화·규제하는 전담 법률을 제정했다. 발사는 국가만의 영역이라는 전제를 깨고, '민간이 발사체를 상업 운영할 수 있다'는 원칙을 세계 최초로 성문화한 것이 CSLA의 역사적 의미다.

| 항목 | 내용 |
|---|---|
| 규제기관 지정 | 상무부(이후 교통부 → 현재 FAA)에 발사 허가·감독 권한 부여 — **7주차**에서 다룬 발사허가 체계의 시원 |
| 민간 발사체 사업 허용 | 민간기업이 자체 소모성발사체(ELV)로 상업위성을 발사할 수 있는 법적 근거 마련 |
| 정책적 배경 | 왕복선 독점에 대한 산업계 불만 + 레이건 행정부의 '작은 정부·시장 활용' 기조가 결합 |
| 한계 | 여전히 왕복선이 주력 수송수단 — CSLA는 '허용'이었을 뿐 실질적 산업 형성은 아직 |

**1986년 챌린저 참사의 전환 효과**

이듬해 챌린저호 폭발(1986.1)로 왕복선 운항이 전면 중단되자, 레이건 대통령은 상업위성의 왕복선 탑재를 사실상 금지하는 정책을 발표 — 이 강제적 '독점 해제'가 역설적으로 민간 ELV 시장(델타·아틀라스·타이탄 상업화)을 실질적으로 열어젖힌 결정적 계기가 되었다.

> **토론 포인트**: 위기(참사)가 규제 완화를 강제한 사례 — 이후 4부에서 볼 안보위협도 유사한 정치적 동력으로 작동한다.

## PART 2 · Ansari X Prize — 상금 하나가 산업을 만들다

| 항목 | 내용 |
|---|---|
| 창설 | Peter Diamandis가 1996년 설립 — 1927년 Orteig Prize(뉴욕-파리 무착륙비행, 린드버그가 수상)를 모델로 삼음 |

| 참가 규모 | 26개 팀, 7개국이 총 약 $1억(상금의 10배)을 자체 투자 — '상금보다 큰 유인 효과'의 전형 |

## PART 2 · 기술혁신의 엔진 — COTS·CRS 앵커계약 (2006~)

XPrize가 '민간도 우주비행이 가능하다'는 기술적 신뢰를 만들었다면, NASA COTS(Commercial Orbital Transportation Services, 2006)는 그 신뢰를 실제 궤도 수송사업으로 전환한 결정적 정책 도구다.

| 구분 | 내용 |
|---|---|
| 계약 방식 | 고정가 마일스톤 지급(Space Act Agreement) — 전통적 원가정산(cost-plus) 방식과 정반대. 목표 달성분만 지급, 초과비용은 기업 부담 |
| 소유권·활용권 | 정부가 설계를 소유하지 않음 — 기업이 개발한 기체로 정부계약 이외 상업시장도 자유롭게 판매 |
| 결과 | SpaceX가 COTS 약 $4억 지원으로 Falcon 9·Dragon 개발, 이후 CRS(2008, $16억 규모 정기 화물수송계약)로 안정적 현금흐름 확보 — 파산 직전 기업을 살린 앵커계약(**11주차** 연계) |
| 파급 | Orbital ATK(현 Northrop Grumman)도 동일 모델로 Antares/Cygnus 개발 — '경쟁 유지형' 복수사업자 육성의 초기 성공사례 |

> 이 모델이 **12주차**(발사서비스 시장경제학)의 학습곡선·발사빈도(LpA) 효과를 처음으로 가능케 한 '초기 물량'의 정책적 기원이다.

## PART 2 · NASA의 역할 전환 — 개발주체에서 '앵커 고객'으로

| | 과거 모델 (Apollo~Shuttle) | 현재 모델 (COTS~Artemis HLS) |
|---|---|---|
| 요구조건 | NASA가 요구사양을 정하고 기업은 하도급 제작만 수행 | NASA는 '요구조건(궤도로 몇 kg)'만 제시, 설계는 기업이 주도 |
| 계약방식 | 설계·소유권·운용 전부 정부 — cost-plus 계약으로 초과비용은 정부가 흡수 | 고정가 마일스톤 — 초과비용은 기업 부담, 정부는 예산 예측가능성 확보 |
| 판매권 | 상업시장 판매 불가 — 정부 전용 자산 | 동일 기체를 상업시장에 자유 판매 — 정부는 '여러 고객 중 하나' |

## PART 3 · 투자 트렌드 — 어디로, 어떻게 자금이 흐르는가

- 발사 서비스는 **딜 건수 비중(17%)보다 자본 비중(30%)이 훨씬 큼** — '소수의 대형 라운드'에 자본이 집중되는 자본집약 산업 특성 (**12주차** 연계)
