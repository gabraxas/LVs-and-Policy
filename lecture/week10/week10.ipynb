## PART 1 · 전략적 무역정책 — 비교우위를 이기는 '선점'의 경제학

**Brander–Spencer 모형 (1985)**

소수의 기업만 생존 가능한 규모의 경제 산업(과점시장)에서는, 정부가 자국 기업에 보조금을 주면 외국 경쟁자의 시장진입 유인 자체를 꺾어 이윤을 자국으로 '이전(profit-shifting)'시킬 수 있다. 비교우위가 없어도, 먼저 움직이고 보조금으로 상대의 진입을 막으면 국가 후생이 개선될 수 있다는 것 — **자유무역의 표준 결론을 뒤집는 이론**이다.

**고전적 실증 사례 — Airbus vs Boeing**: 1970년대 유럽 정부의 대규모 보조금은 Airbus가 Boeing의 '선점자 우위'를 무너뜨리고 복점(duopoly)에 진입하게 했다 — 전략적 무역정책의 교과서적 성공 사례로 꼽힌다.

**발사체 산업에 대한 적용**: 발사체 산업은 막대한 고정비·큰 규모의 경제·소수 공급자라는 Brander-Spencer의 전제 조건을 정확히 충족한다. Ariane, H3, 창정, LVM3는 모두 '시장이 알아서 선택하게' 두지 않고 국가가 먼저 진입시킨 결과다.

## PART 1 · 유치산업론과 학습곡선 — 시간이 만드는 경쟁력

**유치산업론 (Hamilton, List)**

알렉산더 해밀턴과 프리드리히 리스트는 신생 산업이 처음부터 선진국과 자유경쟁하면 규모의 경제·학습효과를 쌓기 전에 도태된다고 주장했다. 초기 보호·지원으로 '유아기'를 넘기면 독자 경쟁력이 생긴다는 논리 — 단, **언제 보호를 걷어낼지에 대한 명확한 출구 조건이 없다**는 것이 고전적 비판이다.

**학습곡선과 TRANSCOST (12주차 연계)**

발사 단가는 누적 생산량이 두 배가 될 때마다 일정 비율로 하락하는 학습곡선을 따른다(TRANSCOST 모델, **12주차**). 발사 횟수 자체가 자산이 되는 구조이기 때문에, 국가가 초기 발사 물량을 앵커계약으로 보장해주는 것은 유치산업 보호의 발사체 버전이라 할 수 있다 — 인도의 저비용 신화, 중국의 140회 목표 모두 이 논리를 깔고 있다.

> **핵심 긴장**: 유치산업 보호가 '언제 끝나는가'라는 문제는 오늘 다룰 다섯 나라 모두에서 반복된다 — 유럽 Ariane 6는 40년째 보호 상태이고, 일본 H3는 절반의 비용 절감을 조건부 자립 목표로 내걸었다.

## PART 1 · 스필오버 — 발사체가 아니라 '능력'을 사는 것이다

발사체 산업 투자의 편익을 로켓 판매 매출만으로 계산하면 거의 모든 나라가 적자다. 그러나 정밀가공·극저온 밸브·복합소재·항법유도제어(GNC)·고에너지 추진 등 발사체 개발 과정에서 축적되는 기술은 항공, 방위산업, 정밀기계, 신소재 산업으로 파급된다 — 이것이 **지식 스필오버(knowledge spillover)**로서, 전략적 무역이론이 보조금을 정당화하는 두 번째 근거(첫째는 이윤이전)다.

| 영역 | 내용 |
|---|---|
| 소재·제조 (극저온·복합재) | 액체수소·메탄 추진계 개발이 조선·에너지 산업의 극저온 저장기술로 파급 (일본·한국 사례) |
| 정밀기계 (GNC·터보펌프) | 유도항법제어와 초정밀 회전기계 기술은 방산·항공기 엔진 산업과 인력을 공유 |
| 다운스트림 (위성·데이터 주권) | 발사 접근성 확보는 통신·관측·항법(PNT) 위성이라는 훨씬 큰 다운스트림 시장의 전제조건 |
| 인력·제도 (고급 엔지니어링 인력) | 대형 시스템 엔지니어링 인력 양성 자체가 국가 산업 역량의 저수지 역할 |

## PART 2 · 로켓과 미사일은 같은 방정식을 공유한다

치올코프스키 로켓방정식(**2주차**)은 우주발사체와 탄도미사일에 동일하게 적용된다. 궤도 투입 능력을 갖췄다는 것은 사실상 대륙간 사거리 투발 능력의 기술적 전제조건을 갖췄다는 뜻이다 — 이것이 발사체 산업을 순수 상업재가 아니라 안보 외부성을 지닌 **이중용도(dual-use)** 기술로 만드는 근본 이유다.

**MTCR (미사일기술통제체제)**: 1987년 결성된 다자 수출통제 체제로, 사거리 300km·탑재중량 500kg을 초과하는 로켓·미사일 관련 기술의 확산을 제한한다. 회원국의 발사체 수출·기술이전이 이 기준에 걸려 제약받는다 — 각국이 '사면 되는' 기술도 국제레짐 때문에 살 수 없는 경우가 생긴다.

**역사적 함의**: 냉전기 소련·미국의 로켓 개발이 대륙간탄도미사일(ICBM) 프로그램에서 파생됐듯(**6주차**), 오늘날 중국 창정·인도 ISRO의 뿌리도 군 미사일 프로그램과 얽혀 있다. 발사체 자립은 곧 '누구의 허락도 없이 궤도에 갈 수 있는 능력'이며, 이는 시장에서 사고파는 상품과는 다른 차원의 국가 능력이다.

## PART 2 · 확증된 우주접근(Assured Access to Space)

미국 법전 10 U.S.C. §2273은 국가안보 탑재체를 "필요할 때 언제든 발사할 수 있는 능력"을 정책 원칙으로 명문화하고, 이를 위해 **최소 2개 이상의 독립적인 발사체 계열을 유지**하도록 규정한다. 즉 세계 최대 발사시장을 가진 미국조차 '단일 공급자 의존'을 법으로 금지한다 — 아무리 저렴하고 신뢰도가 높아도, 공급자가 하나뿐이라는 사실 자체가 안보 리스크라는 논리다.

| 개념 | 내용 |
|---|---|
| 거부 위험(Denial Risk) | 특정 발사체에만 의존하면, 사고·정치적 이유·공급자의 기업 결정만으로 국가안보 임무 전체가 중단될 수 있음 |
| 제도화 사례 — 미국 NSSL | "최소 2개 계열 유지"를 법제화 — Falcon 9의 압도적 경쟁력에도 ULA 등 대안 공급자를 정책적으로 존속시킴 |
| 타국 함의 — 자국 발사체 = 최후 보루 | 자국 발사 능력이 없는 나라는 구조적으로 이 원칙을 충족할 수 없음 — 남에게 100% 의존이 '기본 옵션'이 되는 것 자체가 리스크 |

## PART 2·3 · 우주는 상품이자 상징이다 — 국가위신재로서의 발사체

**국가위신재(Prestige Good) 개념**: 우주개발사에서 국가위신은 반복적으로 등장하는 독립변수다(**6주차** 냉전 우주경쟁). 자체 발사 능력은 "우리는 강대국 클럽의 정회원이다"라는 국제적 신호(signal)를 발신하는 재화다 — 경제성 계산으로는 설명되지 않는 효용이며, 국내 정치적으로는 정권의 정당성·국민통합 서사에도 기여한다.

**발사체 vs 다른 위신재**: 올림픽 유치, 원자력발전, 항공모함 건조와 유사한 논리 구조를 갖는다 — 그러나 발사체는 여기에 '이중용도 안보 기술'이라는 요소가 추가로 결합된, **위신·안보·경제가 삼중으로 겹치는 드문 사례**다. 인도의 화성 탐사선(Mangalyaan)이 서구 대비 압도적 저비용으로 기획된 것도 이 신호 효과를 극대화하려는 전략으로 읽을 수 있다.

## PART 3 · 발전국가론 — 시장이 아니라 관료제가 산업을 고른다

찰머스 존슨(Chalmers Johnson)의 **발전국가(developmental state)** 이론은 전후 일본 통상산업성(MITI)이 시장에 맡기지 않고 전략산업을 직접 지정·육성한 모델을 설명한다. 관료제가 장기 산업 로드맵을 짜고, 민간기업은 그 안에서 경쟁한다 — 자유시장도 순수 사회주의 계획경제도 아닌 제3의 길이다. 일본 H3(정부 개발·민간 상업운용 분리), 한국의 중화학공업화 정책과 동일한 계보에 있다.

| 요소 | 내용 |
|---|---|
| 관료제 조정 | 일본 MITI·JAXA, 인도 IN-SPACe처럼 국가가 산업 방향을 조율하는 전담 기구를 둠 |
| 국가-기업 파트너십 | 위험이 큰 개발단계는 국가가, 상업화는 민간이 맡는 위험 분담 구조가 반복됨 |
| 수출지향 Catch-up | 선진 기술을 빠르게 흡수해 격차를 좁히고, 이후 독자 혁신으로 전환하는 단계적 전략 |

## PART 3 · 국가자본주의와 군민융합 — 중국이라는 제3의 모델

**국가자본주의(State Capitalism)**: 국유기업(CASC)이 시장의 한 참여자가 아니라 산업 전체의 설계자 역할을 겸하는 체제다. 국가가 자본·수요(위성 발주)·규제를 동시에 쥐고 있어, 서구식 정부-기업 분리 모델보다 훨씬 빠른 자원 동원이 가능하다 — 2026년 목표 140회라는 숫자 자체가 시장 수요가 아니라 국가계획에서 역산된 것이 단적인 예다.

**군민융합(军民融合)**: 시진핑 정부가 공식화한 전략으로, 민간(상업) 부문의 혁신 속도를 국방 부문에 흡수하고 국방기술을 민간에 이전하는 양방향 통합을 제도화한다. LandSpace·Galactic Energy 등 '상업' 로켓 기업들이 실제로는 국가 우주굴기 전략의 실행 도구로 기능하는 구조 — 서구의 '민간 대 국가' 이분법이 중국에는 적용되지 않는다.

> **비교 관점**: 미국의 '상업화'(**9주차** COTS)가 국가가 민간에 위임하는 모델이라면, 중국의 군민융합은 국가가 민간의 옷을 입고 직접 조율하는 모델에 가깝다 — 결과(발사 케이던스 급증)는 유사해 보여도 제도적 논리는 다르다.

## PART 5 · 다섯 나라, 다섯 가지 프레임워크 조합

이제 이론을 각국에 적용한다 — 같은 목표(자국 발사체 유지)를 향한 다섯 개의 서로 다른 이론적·제도적 경로.

### 유럽 — 지리적 안배 = 제도화된 전략적 무역정책

**1차 동인: 정치적 자율성 (안보·위신 혼합)**

유럽은 미국·소련(러시아) 어느 쪽에도 발사를 의존하지 않겠다는 냉전기 결정(**6주차**)을 40년째 유지 중이다. Ariane 6의 geo-return 원칙은 Brander-Spencer식 '단일 최적 기업 육성'이 아니라, 27개 회원국의 정치적 결속을 산업 배분으로 구매하는 방식 — 경제학적으로는 비효율적이지만, EU라는 정치체 자체의 존속 비용으로 보면 합리적이다.

**제도적 함의: 발전국가 모델의 부재**

일본·중국과 달리 유럽에는 단일 관료제가 산업을 조율하는 '발전국가'가 없다 — ESA는 주권국가 연합체다. 이것이 유럽이 상업 경쟁력에서 가장 뒤처지는 구조적 이유다: 정치적 합의 비용이 산업 효율성보다 우선시되는 제도 설계이기 때문이다.

### 일본 — 발전국가의 교과서 + 안보 앵커로서의 정찰위성

**1차 동인: 발전국가형 산업정책**

H3의 "정부 개발·민간 상업운용" 분리 구조는 발전국가론이 예측하는 정확한 형태다 — 위험이 큰 개발 단계는 JAXA·정부가, 안정화된 이후 운용은 미쓰비시중공업(MHI)이 맡는다. 비용 절반이라는 명시적 KPI 자체가 관료적 산업 로드맵의 산물이다.

**2차 동인: 안보 앵커 수요의 부재/존재**

일본의 구조적 약점은 확증된 우주접근을 뒷받침할 국내 수요가 얇다는 데 있다. 방위성의 정찰·조기경보 위성 발주가 안정적 앵커 물량이 될 잠재력은 있지만, 미국처럼 법제화된 '최소 2계열' 원칙 수준의 제도적 뒷받침에는 이르지 못했다. Honda·ispace 등 민간의 자생적 진입은 발전국가 모델이 성숙 단계에서 국가 주도→민간 주도로 이양되는 경로를 보여주는 초기 신호로 읽을 수 있다.

### 중국 — 군민융합 + 국가자본주의 = 지경학적 자산으로서의 메가군집

**1차 동인: 국가자본주의적 스케일업**

140회라는 2026년 목표는 시장 수요가 아니라 궈왕·첸판 두 메가군집의 위성 배치 일정에서 역산된 국가계획 숫자다. 국유(CASC)·상업(LandSpace 등) 이중 트랙은 겉보기엔 미국식 상업화와 비슷하지만, 자본·수요·규제를 국가가 동시에 쥔다는 점에서 발전국가 모델보다 더 강한 개입 형태다.

**2차 동인: 안보·경제의 완전한 결합**

군민융합 정책 하에서 '상업 로켓 기업'과 '국방 기술 기반'의 경계가 정책적으로 모호하다. 창정10B의 재사용 성공(2026.7.10)은 단순한 상업적 이정표가 아니라, 미국 의존 없는 독자적 우주굴기 서사의 완성이라는 위신재 성격도 함께 지닌다 — **세 가지 렌즈가 가장 강하게 결합된 사례**다.

### 인도 — 탈식민 기술주권 + 재정 제약하의 최적화

**1차 동인: 탈식민 기술주권(post-colonial techno-nationalism)**

ISRO의 저비용 화성·달 탐사는 국가위신재 이론이 예측하는 '신호 극대화' 전략의 정석이다 — 서구 대비 압도적으로 낮은 비용은 예산 제약의 산물이자 동시에 "제약 속에서도 해낸다"는 서사 자체가 위신 신호로 기능한다.

**2차 동인: 재정 보수성이 만든 전략 선택**

Skyroot의 "빠른 실패 대신 신중한 확장"은 유치산업론이 말하는 '자원이 제한된 후발주자'의 합리적 대응이다 — 미국식 스타트업 자본시장(**9주차**)이 없는 인도에서, 반복 실패를 감당할 벤처 자본이 부족하다는 제약이 오히려 첫 성공에 집중하는 전략을 만든다. IN-SPACe라는 민간 개방 기구 신설은 발전국가에서 시장이양 단계로의 제도적 전환 신호다.

> 2026.7.18 Vikram-1 발사는 "국가 위신재를 국가가 독점하지 않고 민간에 위임하기 시작했다"는 점에서 일본의 초기 발전국가 이양 경로와 유사하지만, 안보 앵커(국방 발주) 없이 순수 상업 동력만으로 진행 중이라는 차이가 있다.

### 러시아 — 경로의존과 매몰비용: 위신재가 부채가 될 때

**1차 동인: 경로의존(path dependency)**

냉전기 압도적 위신재였던 소유즈 체계는 오늘날 정치적 관성과 매몰비용 때문에 유지되는 자산에 가깝다 — 새로운 이론적 정당화(경제·안보·위신)보다는, "한번 만든 능력을 포기하지 못하는" 제도적 관성이 1차 설명변수가 됐다. 안가라조차 로스코스모스 스스로 "실전 배치 전부터 이미 구식"이라 평가한다.

**2차 동인: 제재가 만든 강제 고립**

## PART 6 · 그러나 이 모든 논리는 사후 합리화일 수 있다

지금까지의 프레임워크는 국가 개입을 정당화하는 논리들이다. 그러나 **공공선택이론(Public Choice Theory)**은 정반대 질문을 던진다 — 이 논리들이 실제 정책의 '원인'인가, 아니면 이미 존재하는 지대추구(rent-seeking)를 정당화하는 '사후 서사'에 불과한가?

| 비판 | 내용 |
|---|---|
| 지대추구 — "국가대표 기업"의 함정 | 일단 국가 프로젝트로 지정되면, 실패해도 예산이 계속 투입되는 정치적 잠금 효과 — Ariane 6·안가라의 반복적 지연이 사례 |
| 기회비용 — 다른 산업에 썼다면? | 발사체에 투입된 자금을 순수 비교우위 산업에 투자했을 때의 후생과 비교하는 반사실적(counterfactual) 검증이 거의 이루어지지 않음 |
| 정보 비대칭 — 정부는 '옳은 산업'을 모른다 | Brander-Spencer 모형 자체의 고전적 비판 — 정부가 어느 산업이 전략적으로 옳은지 사전에 알 수 있다는 가정은 비현실적 |

> 위 인터랙티브에서 확인했듯, Brander-Spencer 모형의 결론(보조금이 국가후생을 높인다)은 "정부가 시장구조·비용함수를 정확히 안다"는 강한 가정에 의존한다 — 이 가정이 무너지면 보조금은 순수한 지대추구의 도구로 전락할 수 있다.

## 결론이 아니라 분석 도구를 가져간다

오늘 강의의 목표는 "어느 나라 전략이 옳다"를 판정하는 것이 아니다. 여러분이 **11주차** 이후 비즈니스 전략·사업기획보고서를 쓸 때, 그리고 **14주차** 한국의 우주수송 정책을 평가할 때 던져야 할 질문의 틀을 갖추는 것이다.

1. 이 정책의 1차 동인은 경제·안보·위신 중 무엇이며, 그 비중을 어떻게 검증할 것인가?
2. 정책 수단(보조금·조달의무·수출통제 등)이 그 동인과 논리적으로 정합적인가?
3. 이 개입이 없었을 때의 반사실적 시나리오와 비교해 순편익이 존재하는가?
4. 다른 산업(조선·반도체·전투기·원자력)의 유사 사례에서 무엇을 배울 수 있는가?
