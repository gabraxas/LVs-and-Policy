## 좋은 로켓 ≠ 좋은 사업: 왜 사업모델 분석인가

발사체 산업의 실패 대부분은 기술 실패가 아니라 **사업모델 실패**였다. 훌륭한 엔진을 만들고도 파산한 기업은 많지만, 평범한 기술로 시장을 지배한 기업도 있다. 사업기획보고서가 기술타당성만으로 완성될 수 없는 이유다.

| 위험 | 내용 |
|---|---|
| 죽음의 계곡(Valley of Death) | 발사체 개발은 첫 매출까지 5~10년, 수억 달러가 선행 투입되는 극단적 자본집약 사업 — 기술이 완성되기 전 자금이 고갈되는 것이 가장 흔한 사망 원인 |
| 수요 대비 과잉 공급 | 2018년경 100여 개 소형발사체 기업이 난립했으나 시장 수요는 그 일부만 지탱 가능 — '누구에게, 무엇을, 왜 우리가' 파는지에 대한 답이 없으면 도태 |
| 가격 결정력의 부재 | **12주차**에서 본 것처럼 발사가격은 학습곡선·발사빈도가 결정 — 저빈도 신생기업은 구조적으로 원가 열위에서 출발하므로, 가격이 아닌 다른 가치제안이 필수 |

## 비즈니스 모델 캔버스(BMC): 9블록으로 사업을 한 장에

| 블록 | 질문 |
|---|---|
| 핵심 파트너 (Key Partners) | 누구와 협력해야 하는가? 엔진·소재 공급사? 발사장 운영기관? 정부·우주청과의 관계? |
| 핵심 활동 (Key Activities) | 무엇을 직접 해야 하는가? 설계·제작·발사운영 중 핵심은? |
| 핵심 자원 (Key Resources) | 핵심 자산은 무엇인가? 엔진 기술? 발사장? 인력? 인허가? |
| 가치 제안 (Value Propositions) | 고객의 어떤 문제를 해결하는가? 가격? 일정 확실성? 전용 궤도 투입? 국가 안보적 자립? |
| 고객 관계 (Customer Relationships) | 고객과 어떤 관계를 맺는가? 장기계약? 건별 판매? |
| 채널 (Channels) | 고객에게 어떻게 도달하는가? 직판? 브로커? 정부조달? |
| 고객 세그먼트 (Customer Segments) | 누가 돈을 내는가? 정부(안보·과학)? 상업 위성사업자? 군집위성 운영사? |
| 비용 구조 (Cost Structure) | 무엇에 돈이 드는가? — 개발비(NRC), 생산비, 발사운영비, 발사장·시설 (**12주차** TRANSCOST 구조와 직결) |
| 수익원 (Revenue Streams) | 어떻게 돈을 버는가? — 발사서비스 대금, 정부 개발계약, 부가서비스(위성버스·부품), 자체 서비스 수익 |

## 캔버스 읽는 법: 오른쪽은 '시장', 왼쪽은 '역량', 아래는 '돈'

- **오른쪽 절반 = 시장(Value Side)**: 고객 세그먼트·가치제안·채널·고객관계·수익원. '누가 왜 우리에게 돈을 내는가'에 대한 답. 발사체 기업이 가장 소홀히 하다 실패하는 영역
- **왼쪽 절반 = 역량(Efficiency Side)**: 핵심 파트너·활동·자원·비용구조. '그 가치를 어떻게 만들어내는가'에 대한 답. 수직계열화 vs 외주의 전략적 선택이 여기서 갈림
- **아래 두 블록 = 재무 논리**: 비용구조(왼쪽)와 수익원(오른쪽)의 격차가 곧 손익 — 개발비를 몇 회의 발사로, 몇 년에 걸쳐 회수할 것인가가 발사체 사업 재무계획의 핵심 질문

> **작성 순서 권장**: ① 고객 세그먼트 → ② 가치제안 → ③ 수익원 → ④ 나머지 — '시장에서 출발해 역량으로' (기술에서 출발하지 말 것)

## SpaceX 모델의 전략적 핵심 3가지

**① 정부 앵커계약으로 죽음의 계곡 횡단**

2006~08년 파산 직전의 SpaceX를 살린 것은 NASA COTS 계약(약 4억$ 마일스톤 지급)이었다. 정부가 '개발비를 나눠 내는 첫 고객'이 되어 민간자본만으로 불가능한 초기 개발을 가능케 한 구조 — 이후 CRS 정기 수송계약이 안정적 현금흐름 제공

**② 수직계열화로 비용구조 장악**

엔진부터 발사운영까지 내재화하여 외주 마진을 제거하고 학습곡선 효과를 극대화 — **12주차**에서 본 TRANSCOST 표준 CER이 SpaceX 비용을 과대추정하는 이유가 바로 이 상업적 수직계열화

**③ 자체수요(Starlink)로 수요 리스크 내부화**

발사 수요가 부족하면 스스로 최대 고객이 된다 — Starlink 군집이 Falcon 9의 발사빈도(LpA)를 끌어올려 학습곡선·간접비 효과를 극대화하고, 그 통신매출이 다시 Starship 개발 자금이 되는 순환 구조

## 수익모델과 단위경제: 12주차 비용모형과의 연결

$$\text{발사 1회 마진} = PpF - CpF = PpF - \left(\frac{DEV}{N_a} + MAN_n + OPS_n\right)$$

← **12주차** TRANSCOST 구조

| 개념 | 내용 |
|---|---|
| 매출의 질: 백로그(수주잔고) | 발사체 기업 가치평가의 핵심 지표는 당기 매출이 아니라 계약 백로그 — 몇 년치 발사가 이미 팔려 있는가가 자금조달 협상력을 결정 |
| 현금흐름 타이밍 | 발사 대금은 계약~발사까지 마일스톤 분할 수취가 관행 — 선수금 구조를 잘 설계하면 운전자본 부담이 크게 줄어듦 (재무계획 파트의 단골 논점) |
| 번레이트와 활주로(runway) | 월간 현금소진율(burn rate) 대비 보유현금이 곧 생존 가능 개월수 — '다음 자금조달 이벤트까지 기술 마일스톤을 달성할 수 있는가'가 기획보고서 리스크 파트의 필수 점검항목 |
| 손익분기 발사횟수 | 누적 마진이 개발비(DEV)를 넘어서는 발사횟수 Na*를 명시적으로 제시할 것 — 학습곡선(MANₙ 하락)과 LpA 증가(OPSₙ 하락)를 반영하면 Na*가 앞당겨지는 논리를 정량적으로 보여주는 것이 설득력의 관건 |

## 정부 앵커계약: 민관이 리스크를 나누는 설계

| 구분 | 전통 조달 (Cost-plus) | 앵커계약 (고정가 마일스톤) |
|---|---|---|
| 소유권·설계권 | 정부 소유, 정부 요구사양 | 기업 소유, 기업이 설계 주도 |
| 대금 지급 | 투입비용 + 이윤 보전 | 마일스톤 달성 시 고정액 지급 |

| 정책 효과 | 확실하나 고비용 | 저비용 산업 육성 — 단, 기업 도산 리스크 감수 |

> **왜 국가는 발사체 '기업'을 키워야 하는가**: 국가 주도 개발만으로는 학습곡선·발사빈도·자체수요라는 사업모델 차원의 경쟁력을 만들 수 없다. 우주수송 자립은 결국 '지속가능한 사업모델을 가진 기업'의 존재 여부에 달려 있으며, 정부의 역할은 직접 개발이 아니라 앵커계약·발사허가·보험 제도로 그 사업모델이 작동할 시장환경을 만드는 것이다.

1. SpaceX 모델(수직계열화+자체수요)은 한국 시장 규모에서 재현 가능한가? 불가능하다면 어떤 변형이 필요한가?
2. Rocket Lab의 '발사+우주시스템' 확장은 한국 발사체 기업에게 어떤 시사점을 주는가?
