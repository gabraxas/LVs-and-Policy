## PART 1 · 국제법 기초 — 우주조약과 책임협약

- 모든 국내 발사허가·보험 제도는 두 국제조약에서 출발함
- 발사체는 국경을 넘는 순간부터 국제법의 규율을 받으며, **그 책임은 개인·기업이 아니라 국가가 진다**는 점이 이 분야 법제의 출발점

| 조약 | 핵심 내용 | 발사체 정책상 의미 |
|---|---|---|
| **우주조약**<br>(Outer Space Treaty, 1967) | 우주의 자유이용·비전유, 국가의 국제책임(제6조) — 민간활동도 '발생국(launching State)'이 승인·계속감독 의무 | 각국이 국내법으로 '발사허가' 제도를 두어야 하는 조약상 근거 |
| **책임협약**<br>(Liability Convention, 1972) | 지표면·비행중 항공기 피해는 **절대책임(무과실)**, 우주공간 내 피해(위성간 충돌 등)는 **과실책임** | 발사국이 무과실로도 배상 책임 — 이 리스크를 국내법이 오퍼레이터에게 재분배하는 것이 오늘 강의의 핵심 |

**핵심 구도**

> 국가가 무과실 책임을 지므로 → 국가는 국내법(허가 요건)으로 그 리스크를 관리하고 → 오퍼레이터에게 보험 가입을 의무화해 재무적 리스크를 이전한다.
> 이 3단 구조가 **4부(발사보험)**의 출발점이다 — 아래에서 인터랙티브로 직접 확인한다.

## PART 1 · 판례 — Cosmos 954: 절대책임의 첫 실증 (1978)

- **사건**: 소련 원자로 탑재 정찰위성 Cosmos 954가 궤도이탈, 1978년 캐나다 북서준주에 방사성 잔해 낙하·산포
- **적용 법리**: 책임협약 제2조의 지표면 피해 절대책임 적용 — **소련의 과실 입증 불요**, 피해 발생 사실만으로 책임 성립
- **결과**: 캐나다는 정화비용 약 C$14M을 청구, 최종 C$3M에 합의(1981) — 국가간 청구가 실제로 이행된 사실상 유일한 선례

> **시사점**: 발사국은 '내 잘못이 아니었다'는 항변이 통하지 않는다 — 이것이 모든 발사허가 심사가 극도로 보수적인 이유이며, 왜 정부가 오퍼레이터에게 재무적 책임능력(보험)을 요구하는지의 역사적 뿌리다.

## PART 2 · 국내 법제 — 우주개발진흥법과 우주손해배상법

| 법률 | 핵심 조항 | 내용 |
|---|---|---|
| **우주개발진흥법**<br>(2005 제정) | 제11조 | 우주발사체를 발사하려는 자는 (현) 우주항공청장의 허가를 받아야 함 — 안전성·기술적 타당성·국제의무 준수 여부 등을 심사 |
| **우주손해배상법**<br>(2007 제정) | 무과실책임 | 우주물체 발사로 인한 손해에 대해 발사자는 고의·과실 불문 배상책임(책임협약의 국내 이행 입법) |
| | 책임한도·보험 | 발사자의 배상책임한도를 설정하고, 그 한도 내 보험 가입을 의무화 — 한도 초과분은 정부가 지원 검토 가능(정부-오퍼레이터 리스크 분담 구조) |

**두 법의 관계**

- 우주개발진흥법이 '발사를 허가할 것인가'(**사전 규제**)를 다룬다면, 우주손해배상법은 '잘못됐을 때 누가 배상하는가'(**사후 책임**)를 다룸
- 허가 심사 과정에서 오퍼레이터의 배상능력(보험 증빙)을 확인하는 것이 두 법을 잇는 실무 접점 — 4부에서 다룰 보험이 바로 이 지점에 위치

## PART 4 · 발사보험이 갖는 의미 — 3단 책임 이전 구조

- **11주차** BMC의 '비용구조' 블록에 보험료가 왜 들어가는지는 이 3단 구조를 알아야 이해됨 — 국제법의 무과실 책임이 최종적으로 민간 보험시장의 가격으로 변환되는 경로

| 단계 | 담보 구간 | 내용 |
|---|---|---|
| ① 오퍼레이터 보험 | MPL(최대예상손실)까지 | 발사자가 상업보험시장에서 구매 — 발사허가의 전제조건(financial responsibility) |

## PART 4 · 보험료 산정의 수학 ① — 최대예상손실(MPL)과 Ec

- 제3자 배상책임보험의 필요 보험금액은 임의로 정하지 않음
- FAA·각국 규제기관은 **'기대사상자수(Expected Casualty, Ec)'** 방법론으로 위험을 정량화하고, 이를 화폐가치로 환산해 MPL을 산정

$$E_c = \sum_i P(\text{실패})_i \times A_{c,i} \times \rho_i$$

- $i$ = 비행단계·고장모드별 시나리오, $P(\text{실패})_i$ = 해당 시나리오 발생확률
- $A_{c,i}$ = 낙하 파편의 사상(살상) 면적, $\rho_i$ = 해당 낙하구역의 인구밀도

| 단계 | 산정 절차 | 핵심 변수 |
|---|---|---|
| ① 궤적·파편 모델링 | 비행경로별 잠재 낙하지점과 파편 분포(디브리 필드)를 물리 시뮬레이션으로 산출 | 추력단계·고도·풍향 |
| ② 인구노출 산정 | 낙하구역별 인구밀도(주간·야간 인구모델)를 GIS로 결합 — 해상·무인지역은 ρ≈0 | 인구밀도 데이터 |
| ③ Ec 합산·화폐환산 | 전 시나리오 Ec를 합산 후, 통계적 인명가치(VSL)와 재산피해 추정치를 곱해 MPL(화폐액)로 환산 | VSL, 재산가치 |

> **주의**: MPL은 '가능한 최대손실'이 아니라 **'확률적으로 기대되는' 손실 수준** — 이를 초과하는 손실은 이론상 발생 가능하며, 앞서 본 3단 구조의 ②·③으로 넘어간다 (FAA 공식정의).

## PART 4 · 보험료 산정의 수학 ② — 순보험료에서 영업보험료로

$$\text{순보험료(Pure Premium)} = P(\text{실패}) \times E[\text{손실} \mid \text{실패}]$$

- 실패확률 × 실패 시 기대손실(보험가액 또는 부보금액) — 보험의 가장 기본적인 기대값 원리

$$\text{영업보험료} = \text{순보험료} \times \frac{1 + \text{리스크마진} + \text{사업비율}}{1 - \text{재보험비용률} - \text{이윤율}}$$

- **리스크마진**: 신규기체·희소 데이터의 불확실성 가산 / **사업비율**: 중개수수료·손해사정비 / **이윤율·재보험비용**: 시장 사이클에 연동

| 요인 | 실무 반영 방식 | 예시 |
|---|---|---|
| 실패확률 P | 베이지안 갱신 — 동종 기체·엔진 계열의 과거 발사이력에 사전분포를 결합. 신규 기체는 표본이 적어 신뢰구간이 넓음 → 높은 리스크마진 | 성숙 기체 2~4% vs 신형 기체 6%+ |
| 집적 리스크(Aggregation) | 한 발사에 다수 위성 탑재(라이드셰어) 시 단일 사고로 다수 보험목적물 동시손실 — 상관관계를 반영해 부보한도 산정 | 듀얼 GEO 발사 시 동시 총손실 시나리오 |
| 시장 사이클(Rate on Line) | 요율 = 보험료 ÷ 부보한도. 대형 손실 발생 후에는 이용 가능 용량(capacity)이 줄어 같은 리스크도 요율이 급등(경화시장) | 2023년 손해율 179% → 2024~25년 요율 상승 |

**[인터랙티브] 순보험료 → 영업보험료 계산기** — MPL과 리스크마진·사업비율·재보험비용률·이윤율을 조작해 최종 요율을 시장 통상 요율대(1.5~2.0%)와 비교
