## PART 1 · 궤도 급유 — 두 번째 게임체인저

완전재사용이 'kg당 비용'을 바꾼다면, 궤도 급유는 **'갈 수 있는 곳'**을 바꾼다. LEO에 도달한 Ship은 자체 추진제로는 더 멀리 못 간다 — 탱커 스타십들이 궤도에서 추진제를 채워주면 달·화성급 임무가 열린다.

**① 타깃 Ship 발사** (임무선, 예: HLS가 LEO 대기) → **② 탱커 반복 발사** (HLS 기준 약 10회의 탱커 발사로 추진제 보급) → **③ 도킹·이송** (V3 도킹 포트로 극저온 추진제 선체간 이송) → **④ 고에너지 임무** (달 착륙·화성·군 수송 구상까지)

> **정책적 의미**: NASA Artemis 달 착륙(HLS)이 이 미검증 기술에 의존한다 — 국가 우주계획의 핵심 경로가 한 민간기업의 시험 일정에 묶인 초유의 구조다. 선체간 급유 시연은 2026년 내 목표이나 IFT-12의 재점화 실패로 일정 리스크가 커졌다. IFT-3(2024)에서 동일 선체 내 탱크간 이송까지는 검증됐다.
