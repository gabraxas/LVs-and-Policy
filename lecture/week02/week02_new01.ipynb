## 제트추진의 정의



- 저장된 물질을 분사해 추력을 만들어내는 장치(G.P. Sutton 정의)
- 추진제 질량만을 이용해 고속 분사류를 형성하는 제트엔진(위키피디아 정의)
  - 운동량 보존(${ V }_{ e }$ : 완전팽창 노즐의 배기속도)

     $\left( M - m \right) \times \left( V - \Delta V \right)+ m \times \left( V - { V }_{ e } \right) - M \times V =0$

     $\Longrightarrow \left( M - \dot{ m } \Delta t \right) \times \left( V + \Delta V \right)+\dot{ m } \Delta t \times \left( V - { V }_{ e } \right) - M \times V =0$

     $\Longrightarrow M \frac{ \Delta V }{ \Delta t }=\dot{ m }{ V }_{ e }$

     $ \lim { \Delta t \rightarrow 0} $ 이면,
     
     $\Longrightarrow M \dot{ V }=\dot{ m }{ V }_{ e }$

  - 추력(힘 $F=M \dot{V}$ ) :
  
  $F =\dot{ m }{ V }_{ e }$  =  (단위시간당 분사 질량) × (유효 배기속도)

## 로켓방정식(Rocket Equation)


- 로켓방정식
  - 진공 중 운동방정식
    - $M \frac{ d V }{ d t }= F ={ V }_{ e }\dot{ m }= - { V }_{ e }\frac{ d M }{ d t }$
  - 정리하면
    - $d V = - { V }_{ e }\frac{ d M }{ M }$
  - 만약, 노즐에서의 출구 속도를 $I_{sp}$ x $ g_0 $라 가정하면,

    - $dV = - { I }_{sp}{ g }_{0}\frac{ d M }{ M }. $
  
  - 적분하여 로켓방정식을 얻음
    - $\Delta V ={ I }_{sp}{ g }_{0}\ln {\left( \frac{{ M }_{ i }}{{ M }_{ f }} \right) }$



```python
import numpy as np
import matplotlib.pyplot as plt

# 1. 초기 설정 및 매개변수
m0 = 1000.0       # 로켓 초기 질량 (kg)
mf = 100.0        # 로켓 최종 질량 (연료 소모 후, kg)
v_e = 2500.0      # 배기 속도 (m/s)
total_time = 10.0 # 연소 시간 (초)

# 시간 배열 생성
t = np.linspace(0, total_time, 500)

# 2. 시간에 따른 질량 함수 m(t) (선형 연소 가정)
# m(t) = m0 - (m0 - mf) * (t / total_time)
m_t = m0 - (m0 - mf) * (t / total_time)

# 질량 변화율 dm/dt (음수 값, 연소율)
dm_dt = -(m0 - mf) / total_time

# 3. 로켓 방정식 적분 과정 구현 (가속도 항 -v_e * (1/m) * (dm/dt)를 누적 적분)
# v(t) = -v_e * ln(m(t) / m0)
v_t = -v_e * np.log(m_t / m0)

# 4. 그래프로 시각화 (3개의 서브플롯)
fig, axs = plt.subplots(3, 1, figsize=(8, 10), sharex=True)

# 질량 변화 그래프
axs[0].plot(t, m_t, color='blue', lw=2)
axs[0].set_ylabel('Mass (kg)')
axs[0].set_title('Rocket Mass over Time')
axs[0].grid(True)

# 질량 변화율(dm/dt) 그래프
axs[1].plot(t, np.full_like(t, dm_dt), color='orange', lw=2)
axs[1].set_ylabel('Mass Rate (kg/s)')
axs[1].set_title('Fuel Burn Rate (Constant)')
axs[1].grid(True)

# 속도 변화(적분 결과) 그래프
axs[2].plot(t, v_t, color='green', lw=2)
axs[2].set_xlabel('Time (s)')
axs[2].set_ylabel('Velocity (m/s)')
axs[2].set_title('Integrated Velocity (Delta-v)')
axs[2].grid(True)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. 상수 및 조건 설정
delta_v = 9100.0       # 요구 속도 증분 (m/s)
g0 = 9.80665           # 지구 중력 가속도 (m/s^2)
m_payload = 1000        # 기준 탑재체 중량 (1 ton)

# 비추력 조건 (300초부터 450초까지 50초 간격)
isp_list = [300, 350, 400, 450]

# 가로축: 구조비 (M_final / M_initial)
# 로켓 방정식에 의해 M_final/M_initial 은 exp(-delta_v / (Isp * g0))보다 작아야만 양의 연료 질량이 존재함
structural_ratios = np.linspace(0.01, 0.25, 500)

plt.figure(figsize=(10, 7))

# 2. 비추력 조건별 이륙 중량 계산 및 그래프 플로팅
for isp in isp_list:
    v_e = isp * g0  # 배기 속도 (m/s)

    # 로켓 방정식에서 도출된 총 질량 대 최종 질량의 비 (m0 / mf)
    mass_ratio_req = np.exp(delta_v / v_e)

    # 구조비 = mf / m0 이므로, 요구되는 mass_ratio_req의 역수(mf / m0)보다
    # 실제 구조비가 크거나 같으면 연료를 채울 공간이 없어 이륙 중량이 무한대가 되거나 불가능해짐
    max_structural_ratio = 1.0 / mass_ratio_req

    # 계산 가능한 구조비 영역 필터링 (물리적으로 로켓 성립이 가능한 구간)
    valid_mask = structural_ratios < max_structural_ratio
    x_valid = structural_ratios[valid_mask]

    # 이륙 중량(M_initial) 계산 공식
    # M_initial = M_payload / ( (1 - mass_ratio_req) * (M_final/M_initial) + max_structural_ratio ) 형태로 유도됨
    # 여기서는 직관적으로 탑재체 중량을 기준으로 역산:
    # M_final = x * M_initial
    # M_propellant = M_initial - M_final = (1 - x) * M_initial
    # 실제로는 구조비(x) 내에 구조체(m_struct)와 탑재체(m_payload)가 포함되므로 구조체 계수를 분리해야 정확함.
    # 일반적인 구조비 정의(m_struct / (m_struct + m_propellant))를 적용하여 탑재체 1단위당 이륙중량을 유도:
    m_initial_per_payload = 1.0 / (1.0 - mass_ratio_req * x_valid)

    # 물리적으로 불가능한 영역(이륙중량이 음수가 되는 구간) 제거
    valid_m_mask = m_initial_per_payload > 0

    plt.plot(x_valid[valid_m_mask], m_initial_per_payload[valid_m_mask],
             label=f'Isp = {isp} sec (Max Structural Ratio: {max_structural_ratio:.3f})', lw=2)

# 3. 그래프 스타일링
plt.yscale('log') # 이륙 중량이 급격히 증가하므로 로그 스케일 적용
plt.xlim(0, 0.15)
plt.ylim(1, 1e3)   # 탑재체 중량 대비 1배에서 10,000배까지 표시
plt.xlabel('Structural Ratio ($M_{final} / M_{initial}$)', fontsize=12)
plt.ylabel('Takeoff Weight $ (ton)$ [Log Scale]', fontsize=12)
plt.title('SSTO Takeoff Weight(ton) vs Structural Ratio ($\Delta v$ = 9,100 m/s)', fontsize=14, pad=15)
plt.grid(True, which="both", ls="--", alpha=0.5)
plt.legend(fontsize=10)

plt.tight_layout()
plt.show()
```

## 유효 배기속도(Effective exhaust velocity)


- 유효 배기속도
  - 총 임펄스
  - $I =\int_{0}^{ t } { Tdt }$
  - 총 소모 추진제량
  - ${ M }_{ P }=\int_{0}^{ t } {\dot{ m } d t }$
  - $C =\frac{ d I }{ d { M }_{ p }}=\frac{ T }{\dot{ m }}={ U }_{ e }+\frac{{ A }_{ e }}{\dot{ m }}\left( { P }_{ e } - { P }_{0} \right)$
  - $={ U }_{ e }\left( 1+\frac{{ P }_{ e }{ A }_{ e }}{{ \rho }_{ e }{ U }_{ e }^{2}{ A }_{ e }}\left( 1 - \frac{{ P }_{0}}{{ P }_{ e }} \right) \right)$
  - $={ U }_{ e }\left( 1+\frac{1}{ \gamma { M }_{ e }^{2}}\left( 1 - \frac{{ P }_{0}}{{ P }_{ e }} \right) \right)$
- 출처 : 스탠퍼드 AA284A 강의노트



## 배기속도 — 어떻게 배기속도 Ve를 최대화할 것인가

- 배기속도를 어떻게 최대화할 것인가
- 이론적 최대 배기속도
  - 정압(c.v) 조건에서의 열역학 제1법칙
  - ${h}_{t2}={h}_{t1}+ q ={h}_{ e }+\frac{1}{2}{ U }_{ e }^{2}$
  - 최대 배기속도는 무한팽창 시 얻어짐(${h}_{e}=0$)
  - ${ U }_{ e , max } \sim \sqrt{2{h}_{t2}}=\sqrt{2{ c }_{ p }{ T }_{t2}}=\sqrt{\frac{2 \gamma }{ \gamma - 1} R { T }_{t2}}=\sqrt{\frac{2 \gamma }{ \gamma - 1}\left( \frac{{ R }_{ u }}{{ M }_{ w }} \right){ T }_{t2}}$
  
- 이론적 최대 배기속도는 연소실 온도와 배기가스 분자량에 의해 결정됨

- 등엔트로피 과정 복습
  - 열역학 법칙 정리
    - 열역학 제0법칙
      - A가 C와 열평형, B도 C와 열평형이면 A와 B도 열평형 상태여야 함
    - 열역학 제1법칙
      - 에너지 보존
      - 계는 내부에너지·열·일 세 종류의 에너지만을 가짐
      - $d e = \delta q - \delta w$
      - 단순 물질의 경우 일은 압력일(pressure work)뿐임
      - $\delta w = Pdv$
      - 엔탈피로 표현하면
      - $d h= \delta q + vdP$
- $e$ : 내부에너지
- $h$ : 엔탈피
- $q$ : 열전달량
- $w$ : 일량
- $P$ : 압력
- $v$ : 비체적
- 열역학 복습
    - 열역학 제2법칙
      - 외부 에너지원이 없다면 열은 항상 고온에서 저온으로 흐름

    - 첫 번째 질문) 물질의 온도란 무엇인가?
    - 두 번째 질문) 내부에너지와 엔탈피란 무엇인가?
  - [Ceteris Paribus] 이상기체를 가정
    - 기체 분자의 부피를 무시
    - 완전탄성충돌(운동량 보존)
    - 분자 간 힘의 상호작용 없음
    - 분자의 운동은 뉴턴 법칙을 따르며 무작위적·비상관적임
  - 압력을 다음과 같은 거시적 물성으로 정의함
    - 단위면적당 분자가 벽에 충돌하며 전달하는 운동량 변화

![](https://raw.githubusercontent.com/gabraxas/LVs-and-Policy/main/lecture/week02/assets/s045_img002.gif)

  - 한 변의 길이가 L인 정육면체 용기에 질량 m인 단원자분자 N개가 들어있다고 가정
  - 벽에 작용하는 i번째 기체분자의 힘 = x축 운동량 변화량:
  - $\Delta { p }_{xi}=\left( - m { v }_{xi} \right) - \left( m { v }_{xi} \right)= - 2 m { v }_{xi}$
  - ${ F }_{xi}=\frac{\left| \Delta { p }_{xi} \right|}{ \Delta t }=\frac{2 m { v }_{xi}}{\frac{2 L }{{ v }_{ x }}}=\frac{ m { v }_{xi}^{2}}{ L }$
  - 여기서 벽 충돌 시간간격은 $\Delta t =2 L /{ v }_{ x }$
  - 전체 N개 기체분자가 벽에 작용하는 힘 F는:
  - ${ F }_{ x }=\sum_{ i =1}^{ N } {{ F }_{xi}}=\sum_{ i =1}^{ N } {\frac{\left| \Delta { p }_{xi} \right|}{ \Delta t }}=\frac{ m }{ L }\sum_{ i =1}^{ N } {{ v }_{xi}^{2}}=\frac{ m }{ L } N {\left( { v }_{ x }^{2} \right)}_{avg}$
  - 벽 하나에 작용하는 압력은:
  - $P =\frac{{ F }_{ x }}{{ L }^{2}}=\frac{ m }{{ L }^{3}} N {\left( { v }_{ x }^{2} \right)}_{avg}=\frac{ m }{ V }\left( n { N }_{ A } \right){\left( { v }_{ x }^{2} \right)}_{avg}$
  - $=\frac{ n }{ V } M {\left( { v }_{ x }^{2} \right)}_{avg}$
  - $=\frac{ n M }{ V }{\left( \frac{1}{3}{ v }^{2} \right)}_{avg}$
  - $=\frac{ n M }{3 V }({ v }_{rms}^{2})$
  - $P V = n \left( \frac{ M { v }_{rms}^{2}}{3} \right)$

![](https://raw.githubusercontent.com/gabraxas/LVs-and-Policy/main/lecture/week02/assets/s046_img002.jpg)

- ${ N }_{ A }$ : 아보가드로 수, ~6.02×10²³
  - $P$, $T$, $V$의 이상기체 상태방정식 $P V = nRT$를 이용하면
  - $\frac{ M { v }_{rms}^{2}}{3}= R T$
  - ${ v }_{rms}=\sqrt{\frac{3 R T }{ M }}$
  - 단위 이상기체 분자의 운동에너지는 병진운동 항으로 정의됨
  - ${ K }_{avg}={\left( \frac{1}{2} m { v }^{2} \right)}_{avg}=\frac{1}{2} m {\left( { v }^{2} \right)}_{avg}=\frac{1}{2} m { v }_{rms}^{2}$
  - 온도로 나타내면,
  - ${ K }_{avg}=\frac{1}{2} m \left( \frac{3 R T }{ M } \right)=\frac{3}{2}\left( \frac{ m }{ M } \right) R T =\frac{3}{2}\left( \frac{1}{{ N }_{ A }} \right) R T =\frac{3}{2}\left( \frac{ R }{{ N }_{ A }} \right) T$
  - 볼츠만 상수 ${ k }_{ B }= R /{ N }_{ A }$를 도입하면
  - ${ K }_{avg}=\frac{3}{2}{ k }_{ B } T$
  - 이 식은 온도 T에서 모든 기체분자가 질량에 무관하게 동일한 평균 운동에너지를 가진다는 것을 의미함
  - 앞서 유도한 ${ v }_{ x }^{2}={ v }^{2}/3$ 과정에서, 움직일 수 있는 자유도 f마다 평균 운동에너지 ${ K }_{avg}=1/2{ k }_{ B } T$를 갖는다고 해석할 수 있음
  - 즉 분자운동의 자유도 f를 알면, 분자의 운동에너지(=내부에너지)를 온도의 함수로 나타낼 수 있으며, 그 비례상수는 볼츠만 상수 ${ k }_{ B }$의 절반임

- 출처 : https://in.pinterest.com/pin/chemistry--207024914116267839/
  - 이상기체의 내부에너지 $E$는 분자의 운동에너지 총합과 같음. $n$몰의 기체에는 $n { N }_{ A }$개의 분자가 있으므로
  - $E = n { N }_{ A }{ K }_{avg}= n { N }_{ A }\frac{3}{2}{ k }_{ B } T = n { N }_{ A }\frac{3}{2}\frac{ R }{{ N }_{ A }} T =\frac{3}{2} nRT$
  - 이제 ${ c }_{ v }$와 ${ c }_{ p }$의 정의를 다시 살펴보면
  - ${ c }_{ v } \equiv \frac{1}{ n }\frac{d E }{d T }{\left. ​ \right|}_{ v }=\frac{3}{2} R$
  - ${ c }_{ p } \equiv \frac{1}{ n }\frac{d H }{d T }{\left. ​ \right|}_{ p }=\frac{1}{ n }\frac{d\left( E+PV \right)}{dT}=\frac{1}{n}\frac{dE}{dT}+\frac{1}{n}\frac{d\left( nRT \right)}{dT}={c}_{v}+R=\frac{5}{2}R$
  - 실제 기체의 경우 운동에너지는 자유도 $f$와 다음처럼 관계됨:
  - ${ K }_{avg}= f \frac{1}{2}{ k }_{ B } T$
  - 비열과 그 비는 다음과 같이 됨:
  - ${ c }_{ v }=\frac{ f }{2} R$, 	${ c }_{ p }=\left( \frac{ f }{2}+1 \right) R$
  - $\gamma =\frac{{ c }_{ p }}{{ c }_{ v }}=\frac{ f +2}{ f }$

    - 열용량 증가 — 비열비가 높을수록 정압 조건에서 정적 조건 대비 더 많은 열에너지를 흡수함. 즉 단위질량(또는 몰)당 열에너지 저장 능력이 더 큼
    - 팽창 거동 — 비열비가 클수록 정압 가열 시 기체가 더 잘 팽창하는 경향을 보임. 정압 조건에서 온도를 올리는 데 정적 조건보다 더 많은 에너지가 필요하기 때문임

    - 열역학 사이클의 효율 — 오토(Otto) 사이클, 브레이턴(Brayton) 사이클, 랭킨(Rankine) 사이클 등에서 비열비는 효율에 영향을 미침. 특정 조건에서는 비열비가 높을수록 열효율이 높아질 수 있음

  - 1 bar·298.15K 등온 조건에서, 구성 원소로부터 화합물 1몰이 생성될 때 방출(또는 흡수)되는 에너지, 즉 엔탈피 변화량
  - ${ \Delta }_{ f }{ H }_{CH4}^{0}= - 74.8\frac{ K J }{ mol }= - 4.67\frac{ M J }{ k g }$
  - ${ \Delta }_{ f }{ H }_{H2}^{0}=0\frac{ K J }{ mol }$
- 반응열 ${ \Delta }_{ r }{ H }^{0}$
  - 표준상태(1 bar, 25°C)에서 반응물로부터 생성물이 형성될 때 방출·흡수되는 에너지
  - ${ \Delta }_{ r }{ H }^{0}= \sum {\left[ { n }_{ j }{\left( { \Delta }_{ f }{ H }^{0} \right)}_{ j } \right]}_{products} - \sum {\left[ { n }_{ j }\left( { \Delta }_{ f }{ H }^{0} \right) j \right]}_{reactants}$
- 연소열 ${ \Delta }_{ c }{ H }^{0}$
  - 연료가 산소와 완전연소할 때 방출·흡수되는 에너지
  - ${ H }_{CH4}^{ c }=50.00\frac{ M J }{ k g }$ (LHV)
  - ${ H }_{H2}^{ c }=141.80\frac{ M J }{ k g }$ (LHV)
  - ${ H }_{kerosene}^{ c }=43.00\frac{ M J }{ k g }$ (LHV)
- 생성열
  - 화학에너지(기준 에너지 상태 대비 화학결합에 저장된 에너지)
  - 화학반응의 방출열 추정에 핵심적임
  - 화합물의 생성열 결정 방법
    - 열량계(bomb calorimeter) 측정(반응열로부터)
    - 양자역학적 계산
    - 결합에너지 기반 경험적 추정법
  - 결합가법성(Bond additivity)
    - 결합의 종류와 개수를 파악해 결합별 에너지를 합산
    - 정확도 5% 수준
  - 그룹가법성(Group additivity)
    - 작용기의 종류와 개수를 파악해 그룹별 에너지를 합산
    - 정확도 2% 수준
  - 결합에너지법은 공유결합에 유효하며 고분자에도 적용 가능
  - 복잡한 유기화합물에 대해 상당히 효과적
  - NaCl과 같은 이온성 화합물에는 이 방법을 적용하지 말 것
  - 큐베인(Cubane)처럼 결합각이 크게 뒤틀린 화합물에는 사용하지 말 것
  - 비열·엔트로피 등 다른 물성 예측에도 활용 가능

## 터보펌프 공급식 엔진(Turbopump-fed engine)


- 단일추진제 가스발생기(Monopropellant GG)
  - 초기/원형 동력사이클
  - 성능은 수용 가능한 수준
  - 독립적인 단일추진제 제어로 신뢰성은 높지만, 제3추진제 탑재로 중량이 증가함

  - 가열 가능한 연료량에 따라 추력이 제한됨
  - 일을 뽑아내기 위해 높은 열전달 효율 및/또는 다단터빈이 필요함
