## What is specific heat ratio?

- Specific heat ratio
  - $\gamma$ comes both ${ c }^{ \ast }$ and ${ C }_{ F }$, but the definition of specific heat ratio is
  - $\gamma ={ c }_{ p }/{ c }_{ v }$
  - Why is the universe ruled by $\gamma$?
  - Reason 1) In case of isentropic process and ideal gas EOS, the information is transfer through the speed of sound $a =\sqrt{ \gamma R T }$
  - Reason 2) From the definition of “Temperature” and “Energy of gas”
    - First question) What is the temperature of a matter?
    - Second question) What is the internal energy and enthalpy?
  - $[$Ceteris Paribus] Assume an ideal gas
    - Ignore the volume of gas molecules
    - Perfect elastic collision (conservation of momentum)
    - No force interaction between molecules
    - The motion of molecules follows the Newton’s law and random and irrelevant
  - Define the pressure in a macroscopic property as
    - the change of momentum of molecules owing to the impact of them onto unit area

![](https://raw.githubusercontent.com/gabraxas/LVs-and-Policy/main/lecture/week02/assets/s045_img002.gif)

  - Suppose a cube with a side length of L contains N monoatomic molecules of mass m in a container,
  - The force of one gas molecule (i-th) acting on the wall = the amount of change in momentum in the x-axis:
  - $\Delta { p }_{xi}=\left( - m { v }_{xi} \right) - \left( m { v }_{xi} \right)= - 2 m { v }_{xi}$
  - ${ F }_{xi}=\frac{\left| \Delta { p }_{xi} \right|}{ \Delta t }=\frac{2 m { v }_{xi}}{\frac{2 L }{{ v }_{ x }}}=\frac{ m { v }_{xi}^{2}}{ L }$
  - where, the time interval of impact to the wall is $\Delta t =2 L /{ v }_{ x }$
  - The force $F$ acting of total $N$ gas molecules to a wall becomes:
  - ${ F }_{ x }=\sum_{ i =1}^{ N } {{ F }_{xi}}=\sum_{ i =1}^{ N } {\frac{\left| \Delta { p }_{xi} \right|}{ \Delta t }}=\frac{ m }{ L }\sum_{ i =1}^{ N } {{ v }_{xi}^{2}}=\frac{ m }{ L } N {\left( { v }_{ x }^{2} \right)}_{avg}$
  - The pressure acting on one wall is :
  - $P =\frac{{ F }_{ x }}{{ L }^{2}}=\frac{ m }{{ L }^{3}} N {\left( { v }_{ x }^{2} \right)}_{avg}=\frac{ m }{ V }\left( n { N }_{ A } \right){\left( { v }_{ x }^{2} \right)}_{avg}$
  - $=\frac{ n }{ V } M {\left( { v }_{ x }^{2} \right)}_{avg}$
  - $=\frac{ n M }{ V }{\left( \frac{1}{3}{ v }^{2} \right)}_{avg}$
  - $=\frac{ n M }{3 V }({ v }_{rms}^{2})$
  - $P V = n \left( \frac{ M { v }_{rms}^{2}}{3} \right)$

![](https://raw.githubusercontent.com/gabraxas/LVs-and-Policy/main/lecture/week02/assets/s046_img002.jpg)

- ${ N }_{ A }$ : Avogador’s number, ~6.02x1023
  - We have another correlation between $P , TandV$ of EOS, $P V = nRT$
  - $\frac{ M { v }_{rms}^{2}}{3}= R T$
  - ${ v }_{rms}=\sqrt{\frac{3 R T }{ M }}$
  - The kinetic energy of unit ideal gas molecular can be defined as the term of translational motion
  - ${ K }_{avg}={\left( \frac{1}{2} m { v }^{2} \right)}_{avg}=\frac{1}{2} m {\left( { v }^{2} \right)}_{avg}=\frac{1}{2} m { v }_{rms}^{2}$
  - In terms of temperature,
  - ${ K }_{avg}=\frac{1}{2} m \left( \frac{3 R T }{ M } \right)=\frac{3}{2}\left( \frac{ m }{ M } \right) R T =\frac{3}{2}\left( \frac{1}{{ N }_{ A }} \right) R T =\frac{3}{2}\left( \frac{ R }{{ N }_{ A }} \right) T$
  - By introducing the Boltzmann constant, ${ k }_{ B }= R /{ N }_{ A }$
  - ${ K }_{avg}=\frac{3}{2}{ k }_{ B } T$
  - This equation means that at temperature T, all gas molecules have the same average kinetic energy regardless of their mass
  - From the process of deriving the preceding ${ v }_{ x }^{2}={ v }^{2}/3$, it can be interpreted as having an average kinetic energy of ${ K }_{avg}=1/2{ k }_{ B } T$ for every degree of freedom f that can move
  - Thus, if we know the degree of freedom of molecular motion f, we can express the kinetic energy (= internal energy) of the molecule as a function of temperature, and its constant of proportionality is half of the Boltzmann constant ${ k }_{ B }$

- source : https://in.pinterest.com/pin/chemistry--207024914116267839/
  - For ideal gas, the internal energy $E$ is equal to the kinetic energy of molecules. And there are $n { N }_{ A }$ molecules in $n$ mole of gas
  - $E = n { N }_{ A }{ K }_{avg}= n { N }_{ A }\frac{3}{2}{ k }_{ B } T = n { N }_{ A }\frac{3}{2}\frac{ R }{{ N }_{ A }} T =\frac{3}{2} nRT$
  - Now, let’s recall the definition of ${ c }_{ v }$ and ${ c }_{ p }$
  - ${ c }_{ v } \equiv \frac{1}{ n }\frac{d E }{d T }{\left. ​ \right|}_{ v }=\frac{3}{2} R$
  - ${ c }_{ p } \equiv \frac{1}{ n }\frac{d H }{d T }{\left. ​ \right|}_{ p }=\frac{1}{ n }\frac{d\left( E+PV \right)}{dT}=\frac{1}{n}\frac{dE}{dT}+\frac{1}{n}\frac{d\left( nRT \right)}{dT}={c}_{v}+R=\frac{5}{2}R$
  - For real gas, the kinetic energy relates to the degree of freedom of $f$ as follow:
  - ${ K }_{avg}= f \frac{1}{2}{ k }_{ B } T$
  - And the specific heat and their ratio becomes:
  - ${ c }_{ v }=\frac{ f }{2} R$, 	${ c }_{ p }=\left( \frac{ f }{2}+1 \right) R$
  - $\gamma =\frac{{ c }_{ p }}{{ c }_{ v }}=\frac{ f +2}{ f }$

![](https://raw.githubusercontent.com/gabraxas/LVs-and-Policy/main/lecture/week02/assets/s048_img002.jpg)

  - Larger specific heat ratio implies :
    - Increased Thermal Capacity: A higher specific heat ratio indicates that the gas can absorb more heat energy at constant pressure compared to its capacity at constant volume. This means that the gas has a greater capacity to store thermal energy per unit mass or per mole.
    - Expansion Behavior: A larger specific heat ratio implies that the gas is more likely to expand when heated at constant pressure. This is because more energy is required to raise its temperature at constant pressure compared to constant volume.
    - Adiabatic Processes: In adiabatic processes, a higher specific heat ratio leads to a more significant change in temperature for a given change in pressure. This is because the relationship between temperature and pressure in adiabatic processes depends on the specific heat ratio.
    - Speed of Sound: A larger specific heat ratio typically corresponds to a higher speed of sound in the gas.
    - Efficiency in Thermodynamic Processes: In thermodynamic cycles such as the Otto cycle, Brayton cycle, or Rankine cycle, the specific heat ratio influences the efficiency of these processes. Higher specific heat ratios can lead to higher thermal efficiencies under certain conditions.

## Combustion instability

- Occurs when different factors in the vehicle and engine systems trigger pressure oscillations of various frequencies and amplitudes.
- Depending on the frequency of pressure oscillations, the instability is classified into three categories
  - – Chugging (10-400 Hz) – Linked to interactions with the propellant feed system or vehicle and the combustion chamber.
  - – Buzzing (400-1000 Hz) – Linked to interactions with the engine structure, mixture ratio, and propellant feed system resonances
  - – Screaming (>1000 Hz) – Linked to the combustion process and acoustic resonance properties in the combustion chamber.
- Modes of Combustion Instability
  - Tangential Mode
  - Radial Mode
  - Combined Mode
  - Longitude mode

- The Impulse Turbine uses only kinetic energy of working gas to rotate the turbine rotor.
- The Reaction Turbine uses both pressure and kinetic energy of working medium to rotate the turbine rotor.
- Impulse turbines are typically used for high-head, low-flow applications such as hydroelectric power plants, where the fluid is delivered under high pressure.
- Reaction turbines are typically used for low-head, high-flow applications such as hydropower plants, where the fluid is delivered at a lower pressure.
- Key decision parameter for LRE design process

`4.1 Test of LREs`

## Key Terms

- Development test :
  - – To achieve design maturity, demonstrate capability, and to reduce risk to the qualification program
  - – Validate new design concepts or the application of proven concepts and techniques to a new configuration,
  - – Assist in the evolution of designs from the conceptual phase to the operational phase,
  - – Validate design changes,
  - – Reduce the risk involved in committing designs to the fabrication of qualification and flight hardware,
  - – Develop and validate qualification and acceptance test procedures,
  - – Investigate problems or concerns that arise after successful qualification,
- Qualification test :
  - – Demonstrate that the design, manufacturing process, and acceptance program produce hardware/software that meet specification requirements with adequate margin to accommodate multiple rework and test cycles,
  - – In addition, the qualification tests should validate the planned acceptance program, including test techniques, procedures, equipment, instrumentation, and software.
- Acceptance test :
  - –  To demonstrate the acceptability of each deliverable item to meet performance specification and demonstrate error-free workmanship in manufacturing
  - – Stress screen items to precipitate incipient failures due to latent defects in parts, processes, materials, and workmanship,
  - – Component acceptance testing at the bench level serves to reduce risk for engine acceptance testing, but it may not simulate the engine environments adequately.
- source : AIAA LRE Course

## Features of Engine Combustion Test Facilities

- High Level Difference (Building Height [Lower Step])
- Safe discharge purpose of engine wake flame
- Engine Ground Combustion Test Facility Building: 48.4m[28.7m]
- Engine High Altitude Combustion Test Facility Building: 58.0m(38.3m)
- 3rd Engine Combustion Test Facility Building: 35.1m[19.7m]
- (2) Barriers greater than 1 m
- Protective fence for safety, fuel and fire  prevents flame propagation with oxidizing agent
- (3) Thrust Transmission Structure
- No shaking or bending when transmitting propulsion
- (4) Thrust Compensation (Measurement)
- No bending
