## 열역학Thermodynamics 시대의 서막

- 1610, 갈릴레오 갈릴레오(Galileo Galilei) : 온도를 측정하는 도구 사용. 당시에는 온도라는 개념이 정리되지 않았음. 뜨거움과 차가움의 변화 정도를 나타내는 도구
- 1630, 페르디난드 2세(Ferdinand II) : 알코올 온도계
- 1670, 로버트 보일(Robert Boil) : 보일의 법칙. P V = c
- 1770, 조세프 블랙(Joseph Black) : 열소이론(Caloric theory). 열은 물질의 일종. 열소가 뜨거운 물체에서 차가운 물로 이동. 블랙은 열량보존 개념으로부터 물질보존 개념을 유추하여 칼로릭의 존재를 주장함. 문제점 1) 열소가 물질이라면 무게가 있어야 하는데, 무게가 있다면 온도에 따라 물질의 무게가 변경되어야 한다. 하지만 관측 결과는 그러하지 않다. 문제점 2) 열소는 열을 수반하는 모든 과정에서 보존된다는 주장
- 1783, 라부아지에(A.L. Lavoisier) : 질량 보전의 법칙. 화학 반응 전후에 있어서 반응전 물질의 총질량과 생성물 총질량은 변함이 없다. 플로지스톤 이론 배척. 연소는 연소물이 산소와 결합하는 것. 징세 청부인/화학 감독관으로 프랑스에서 부를 축적후 매일 아침 6~9시, 저녁 7~10시까지 화학실험에 매진.
- 1789, 럼포드(B.T. Rumford) : 열소이론 반박. 물통에 대포알을 넣고 드릴날로 마찰 후 통 안의 물이 모두 끓어 사라지고 없지만, 열소는 찾지 못했음. 열은 질량을 갖지 않는다는 것이 정설화됨. 그는 오랜 실험을 통해 열은 물질을 구성하는 분자들의 운동에 의한 것이라는 결론을 내림
- 1799, 프루스트(J.R. Proust) : 일정 성분비의 법칙. 한 화합물(e.g H2O)을 구성하는 각 성분 원소(H : O)들의 질량 비가 일정(1:8)하다는 법
- 1802, 게이뤼삭(J.L. Gay-Lussac) : 온도의 증가에 따라 기체의 부피가 선형적으로 증가. 절대온도 (부피가 0이 되는 온도)개념. 게이뤼삭의 기체의 제2법칙, 3법칙 발표. 기체가 열을 얻으면 팽창, 팽창이 불가능한 상황이라면 압력이 높아짐
- 1803, 존 돌턴(John Dalton) : 모든 기체는 온도의 증가에 따라 부피가 균일하게 증가하는 사실 발견. 당시 공기는 적어도 1/5의 산소와 4/5의 질소의 혼합물이라는 사실이 알려져 있었음. 돌턴은 기체의 혼합물이 어떻게 하나의 압력을 만드는가에 대해서 궁금해함. 볼턴은 앞서 게이뤼삭의 생각과는 반대로 기체 압력과 부피에 따라 온도는 어떠한 영향을 받는지 궁금해하였음. 이를 해결하기 위해 분압이라는 개념을 도입(분압의 법칙)
- 1840, 줄리우스 폰 마이어(Julius von Mayer) : 체내에서 산화되는 음식(연료)가 체온을 유지하거나 운동을 하는 몸의 활동과 관계가 있다고 생각함. 음식을 섭취한 몸이 다른 물체에게 운동과 열을 전달하는 것을 통해, 에너지를 만들어 내거나 없애지 못하지만 여러 가지 형태로 변환하다는 열역학 1법칙을 처음 밝힘
- 1884, 제임스 줄(J. Joule) : 줄의 법 발견. 열과 일이 동등하다는 것을 실험으로 증명하면서, 일을 해줘서 열이 생성되는 것으로부터 물질 내부에 열소가 포함되어 있는 것이 아님을 주장. 열소이론 소멸.
- 1765, 제임스 와트(James Watt) : 증기기관의 개선
- 1824, 사디 카르노(Sadi Carnot) : 열기관에 대한 연구를 통해 가역과정(reversible process) 개념을 처음 사용하여 가역적 순환공정을 고안함. 이를 통해 카르노 사이클(Carnot Cycle) 제안
  - 사디 카로노의 아버지 라자르 카르노(Lazare Carnot)는 나폴레옹 백일천하 때 내무장관이었으며, 나폴레옹이 워털루에서 패배하고 유배형에 처해지면서, 아들인 사디 카로노는 군대에 끌려가게 됨
  - 당시 영국을 시작으로 증기기관을 활용하여 산업화가 시작되면서, 프랑스에서도 증기기관에 대한 관심이 많아지게 되었음  증기기관에 관심이 많았던 카르노는 당시 증기기관의 효율을 높이기 위해서는 마찰을 비롯한 설계상의 결함을 배제하고 가상의 완벽한 ’열기관’을 생각하여 이론적인 연구를 시작함
  - 그는 열소이론(Caloric)을 기초로, 칼로릭이 물레방아를 돌리는 물과 같이, 엔진의 운동량은 엔진을 지나가는 칼로릭의 양에 전적으로 의존한다는 생각을 고안함.
  - 즉, 칼로릭을 운반하는 물질이 무엇이든 상관없이, 열원과 열침(heat sink) 사이의 온도차가 크면 클수록 칼로릭의 흐름이 커진다고 생각했음
  - 1830년 7월 혁명으로 연구가 단절되고, 2년 후 그는 콜레라로 36세에 사망함.
- “열은 언제나 퍼져나가고 평준화된다.” by Boerhaave (17th century)
- 뉴턴 역학 vs. 열역학
    - 뉴턴 역학에서 운동과 질량 등은 모두 열역학에 비하여 상대적으로 계측이 좀 더 용이한 물리량(e.g. 무게, 질량, 속도, 가속도 등)을 가진 반면,
    - 열역학에서 온도를 정량적인 형태로 계측하는 것은 18세기에 들어서서야 가능해졌으며, 심지어 19세기 중반에 영국의 물리학자 Thomson (후에 Lord Kelvin)에 의해 절대온도가 발명됨

## 열 기관heat cycle


> 원본 슬라이드에는 카르노 사이클 애니메이션(동영상) 3개가 삽입되어 있으나 노트북 형식에서는 재생할 수 없다 — 다음 절의 카르노 사이클 정적 다이어그램으로 대체 확인.

- 카르노의 생각 :
  - 두 열원 T1과 T2가 있을 때,
  - 가장 많은 일을 해내기 위한 열기관은 일을 해내기 위한 가스가,
  - 고온부로부터 에너지를 전달받고,
  - 이 에너지를 이용하여 일을 하고
  - 저온부로 열이 빠져나가야 함
  - 이 때, 최대로 효과적인 열기관을 구성하는 방법은
    - 열원 T1으로부터 작동 기체로 열을 전달받는 과정에서 작동 기체의 온도는 T1일 때가 최대임. 즉 등온팽창 과정
    - 작동기체가 외부로 일을 하기 위해 팽창하는 과정에서 외부로 불필요한 열전달은 없이 오직 부피 팽창만 함. 팽창은 기체의 온도가 T2가 될 때까지 수행. 즉, 단열팽창 과정
    - 작동기체가 일을 해준 후 다시 원래지점으로 돌아가서 사이클을 만들기 위해서는 저온부 T2로 열을 전달하고 압축되어야 함. 이 때 작동가스 온도가 최소가 되는 것은 T2일 때이며, 만약 2번에서 더 높은 온도까지 팽창만 하면 손실이 발생함.  즉, 등온압축 과정
    - 다시 온도가 T1으로 상승하여 1번 과정에서 최대로 열전달을 받기 위해서 T1까지 압축하여 부피는 감소하고 압력은 상승. 이 압축과정이 가장 효율적이 되기 위해서는 외부와의 열전달이 없어야 함. 즉, 단열압축 과정

## 클라우지우스의 등장과 카르노 이론의 모순

- Rudolf Clausius가 27세의 고등학교 교사였던 1849년 Clausius가 프랑스 과학자 Sadi Carnot의 열 이론에 관한 논문을 읽은 후 열에 대한 Carnot의 생각 중 일부가 옳다고 생각했지만 열(or 칼로릭)이 항상 보존된다는 생각에는 동의하지 않았음
- 1850년에 클라우시우스는 열이 일로, 일이 열로 전환될 수 있다는 (현재 우리가옳다고 믿고 있는) 이론을 발표했음
- 또한 Clausius는 그가 내부 일이라고 부르는 것과 내부 열을 결합한 새로운 용어를 추가했으며(이제 우리는 내부 에너지라고 부름), 이 용어로 Clausius는 에너지가 보존된다는 열역학 제1법칙의 최초의 완전한 버전을 만들었음
- Clausius는 그 후 13년 동안 ”에너지”라는 단어를 사용하지 않았지만(” 에너지 ” 라는 용어가 1850년대와 1860년대에 처음으로 대중화됨에 따라) Clausius가 만든 방정식은 여전히 동일한 문자와 기호로 첫 번째 법칙을 설명하는 데 사용됨

## 열역학 제2법칙

- 1865년 클라우시우스는 “엔트로피(entropy)”라는 단어가 변형을 뜻하는 그리스어에서 유래했으며 “에너지(energy)”라는 단어와 최대한 유사하도록 엔트로피라는 단어를 의도적으로 만들었음

![](https://raw.githubusercontent.com/gabraxas/LVs-and-Policy/main/lecture/week02/assets/s042_img040.png)


![](https://raw.githubusercontent.com/gabraxas/LVs-and-Policy/main/lecture/week02/assets/s043_img041.png)


![](https://raw.githubusercontent.com/gabraxas/LVs-and-Policy/main/lecture/week02/assets/s044_img042.jpg)

- 절대 온도 T인 열역학적 계가 ΔQ만큼의 열을 흡수했을 때 엔트로피 변화량
- A계는 Q 만큼의 열을 잃었으므로 A계의 엔트로피 변화량: ΔSA = -Q∕T₁  B계는 Q 만큼의 열을 얻었으므로 B계의 엔트로피 변화량: ΔSB = Q∕T₂
- 전체 계(A+B)의 엔트로피 변화량은 A계와 B계의 엔트로피 변화량을 합.
- ΔS전체 = ΔSA + ΔSB = -Q∕T₁ + Q∕T₂ = Q (1∕T₂ − 1∕T₁)
- 그런데 초기 온도는 T₁ > T₂ 이었으므로, 이 ΔS전체 = Q (1∕T₂ − 1∕T₁) > 0이 된다. 즉, 엔트로피가 증가한 것이다.
- 한편 클라우디우스의 큰 비판자 중 한 명인 윌리엄 톰슨은 상응하는 방정식이 없었지만 자신만의 “제2법칙”을 만들고 있었음.
- 1852년에 Thomson은 “더 낮은 온도에서 열이 한 물체에서 다른 물체로 전달될 때 인간이 사용할 수 있는 기계적 에너지의 낭비가 항상 있다”라고 기술하였음.
- 1862년까지 Thomson은 “열역학의 두 번째 대법칙은 자연에서 비가역적인 작용의 특정 원리를 포함함. 따라서 역학적 에너지는 파괴할 수 없지만 물질 우주를 통한 열의 점진적인 증가 및 확산, 운동의 중단 및 위치 에너지의 소진을 일으키는 보편적인 소멸 경향이 있음”을 보여줌.
- 1865년에 Clausius가 자신의 함수 엔트로피를 명명했을 때, Clausius는 엔트로피에 대한 그의 방정식과 이론이 “에너지 낭비(wasting energy)”에 대한 Thomson의 아이디어와 완벽하게 일치한다는 것을 깨달았음!
- 즉, 돌이킬 수 없는(irreversible) 자연의 작용(action of nature)에 상응하는 방정식이 생겼음.
